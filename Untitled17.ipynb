```python
#
# Gravitational Wave Injection & Analysis Pipeline (SNR-Only Version)
#
# This script uses a scientifically robust method to:
# 1. Fetch real LIGO noise.
# 2. Measure the noise characteristics (PSD) of the real data.
# 3. Inject a simulated signal scaled accurately to the measured noise.
# 4. Perform a full analysis including matched filtering and Q-transform.
# 5. Provide complete visualizations at each major step.
#

import matplotlib.pyplot as plt
import matplotlib as mpl
from pycbc.types import TimeSeries
from pycbc import detector
from pycbc.psd import interpolate, inverse_spectrum_truncation
from pycbc.waveform import get_td_waveform
from pycbc.filter import sigma, matched_filter, resample_to_delta_t, highpass
import pywt  # For wavelet transforms
from gwpy import timeseries
import numpy as np
# Fix datetime imports




# -----------------------------------------------------------------------------
# SECTION 1: CORRECTED SIGNAL INJECTION FUNCTION
# -----------------------------------------------------------------------------

def inject_signal_corrected(real_noise_data, real_psd, params):
    """
    Injects a simulated signal into a provided timeseries of real LIGO noise,
    using a provided real PSD to accurately scale for the target SNR.
    """
    print("--> Generating simulated waveform...")
    det = detector.Detector('H1')

    hp, hc = get_td_waveform(
        mass1=params['mass1'], mass2=params['mass2'], spin1z=params['spin1z'], spin2z=params['spin2z'],
        inclination=params['inclination'], f_lower=params['f_lower'], approximant=params['approximant'],
        delta_t=1.0 / params['sample_rate']
    )

    h = det.project_wave(hp, hc, params['ra'], params['dec'], 0)

    # Resize the waveform to match the PSD's duration BEFORE calculating the scaling factor.
    h.resize(len(real_noise_data))

    print("--> Waveform generated. Calculating distance for target SNR using REAL PSD...")
    distance_scale_factor = sigma(h, psd=real_psd, low_frequency_cutoff=params['f_lower']) / params['snr']
    h /= distance_scale_factor

    injection_time = float(real_noise_data.start_time) + real_noise_data.duration / 2
    h.start_time += injection_time

    print(f"--> Injecting signal at GPS time {injection_time:.2f} with target SNR={params['snr']}")
    dat_inj = real_noise_data.add_into(h)

    return dat_inj, h, injection_time

# -----------------------------------------------------------------------------
# SECTION 1.5: USER INPUT FUNCTIONS
# -----------------------------------------------------------------------------

def get_initial_parameters():
    """
    Get user input for binary black hole parameters and SNR.
    """
    print("\n=== Binary Black Hole Parameter Input ===")
    print("Please enter the following parameters (press Enter for defaults):")

    try:
        # Get mass inputs with validation
        mass1_input = input("Mass of first black hole in solar masses [default=36]: ")
        mass1 = float(mass1_input) if mass1_input.strip() else 36
        if mass1 <= 0:
            print("Mass must be positive. Using default value of 36.")
            mass1 = 36

        mass2_input = input("Mass of second black hole in solar masses [default=29]: ")
        mass2 = float(mass2_input) if mass2_input.strip() else 29
        if mass2 <= 0:
            print("Mass must be positive. Using default value of 29.")
            mass2 = 29

        # Get SNR input
        snr_input = input("Signal-to-noise ratio (SNR) [default=1000]: ")
        snr = float(snr_input) if snr_input.strip() else 1000
        if snr <= 0:
            print("SNR must be positive. Using default value of 1000.")
            snr = 1000

        return {
            'mass1': mass1,
            'mass2': mass2,
            'snr': snr,
            # Keep other parameters the same
            'spin1z': 0,
            'spin2z': 0,
            'inclination': 0,
            'ra': 2.3,
            'dec': -1.2,
            'f_lower': 20,
            'approximant': 'IMRPhenomXPHM',
            'sample_rate': 4096
        }
    except ValueError as e:
        print(f"Invalid input: {e}. Using default parameters.")
        return {
            'mass1': 36,
            'mass2': 29,
            'snr': 1000,
            'spin1z': 0,
            'spin2z': 0,
            'inclination': 0,
            'ra': 2.3,
            'dec': -1.2,
            'f_lower': 20,
            'approximant': 'IMRPhenomXPHM',
            'sample_rate': 4096
        }

# -----------------------------------------------------------------------------
# SECTION 1.6: SNR-ONLY ANALYSIS WINDOW FUNCTION (NEW)
# -----------------------------------------------------------------------------

def find_optimal_analysis_window(white_data_bp, actual_merger_time, params):
    """SNR-only wrapper function for backward compatibility"""
    # Extract only SNR from params
    snr = params.get('snr', 30)  # Default to 30 if not found
    return find_optimal_analysis_window_snr_only(white_data_bp, actual_merger_time, snr)

def find_optimal_analysis_window_snr_only(white_data_bp, actual_merger_time, snr):
    """
    SNR-only window analysis for gravitational wave signals without assuming masses.
    Uses three distinct methods that all adapt solely based on SNR.

    Parameters:
        white_data_bp: Whitened and bandpassed data
        actual_merger_time: Time of merger peak
        snr: Estimated signal-to-noise ratio

    Returns:
        window_before, window_after (in seconds)
    """
    print("\n--> Finding optimal analysis window using SNR-only approach...")

    # Get the data as numpy arrays
    times = white_data_bp.sample_times.numpy()
    values = white_data_bp.numpy()

    snr_factor = min(1.0, np.log10(max(10, snr)) / np.log10(1000))
    print(f"     SNR = {snr}, SNR scaling factor = {snr_factor:.3f}")
    # Basic validation to ensure we have enough data
    if len(times) < 100:
        print("     WARNING: Not enough data points for robust window analysis. Using defaults.")
        return 0.10, 0.003

    # Find the merger time index and ensure it's valid
    merger_idx = np.argmin(np.abs(times - actual_merger_time))
    if merger_idx <= 0 or merger_idx >= len(times) - 1:
        print("     WARNING: Merger time outside valid data range. Using defaults.")
        return 0.10, 0.003

    print("     Method 1: Applying SNR-based window estimation...")

    # Pure SNR-based window without mass assumptions
    # Based on empirical observations of detectability:
    # - At SNR=10: ~50ms of inspiral is visible
    # - At SNR=100: ~150ms of inspiral is visible
    # - At SNR=1000: ~300ms of inspiral is visible

    # Use logarithmic scaling with SNR
    min_window = 0.03  # Minimum 30ms window at very low SNR
    max_window = 0.25  # Maximum 350ms window at very high SNR

    # Calculate window sizes based on SNR
    snr_window_before = min_window + (max_window - min_window) * snr_factor

    # After-merger window also scales with SNR but over smaller range
    min_after = 0.002  # Minimum 2ms after window
    max_after = 0.010  # Maximum 10ms after window
    snr_window_after = min_after + (max_after - min_after) * snr_factor
    # Rest of the function continues as before...
    # [envelope detection, spectral analysis, etc.]

    # When setting method weights, replace with regime-specific weights:
    if snr > 100:  # High SNR - data-driven methods more reliable
        print("     High SNR detected: Weighting envelope detection more heavily")
        weights = [0.15, 0.55, 0.3]  # [Theoretical, Envelope, Spectral]
    elif snr > 30:  # Medium SNR - balanced approach
        print("     Medium SNR detected: Using balanced weights")
        weights = [0.25, 0.45, 0.3]
    else:  # Low SNR - rely more on theoretical scaling
        print("     Low SNR detected: Weighting theoretical method more heavily")
        weights = [0.2, 0.6, 0.2]

    # Calculate weighted average as before, but with updated weights and methods
    # window_before = weights[0] * snr_window_before + weights[1] * env_window_before + weights[2] * spec_window_before
    print(f"     SNR-based window: {snr_window_before:.4f}s before merger")


    # print(f"     Using weighted average: {weights[0]:.1f}*Theory + {weights[1]:.1f}*envelope + {weights[2]:.1f}*spectral")



    # --------------------------------------------------------------------------------
    # METHOD 2: ENVELOPE-BASED DETECTION (SNR-OPTIMIZED)
    # --------------------------------------------------------------------------------
    print("     Method 2: Analyzing signal envelope with SNR-optimized threshold...")

    try:
        from scipy.signal import hilbert
        analytic_signal = hilbert(values)
        amplitude_envelope = np.abs(analytic_signal)

        # Smooth the envelope - use wider smoothing for lower SNR
        # Lower SNR requires more smoothing to overcome noise
        smooth_window_seconds = 0.006 - 0.004 * snr_factor  # 6ms at low SNR, 2ms at high SNR
        window_len = max(5, int(white_data_bp.sample_rate * smooth_window_seconds))
        if window_len % 2 == 0:
            window_len += 1  # Ensure odd window length

        smooth_envelope = np.convolve(amplitude_envelope,
                                    np.ones(window_len)/window_len,
                                    mode='same')

        # Normalize envelope
        smooth_envelope = smooth_envelope / np.max(smooth_envelope)

        # Calculate background noise level from first 20% of data
        noise_idx = int(0.2 * len(smooth_envelope))
        bg_values = smooth_envelope[:noise_idx]

        # SNR-adaptive threshold:
        # - Low SNR: Higher threshold (more standard deviations above noise)
        # - High SNR: Lower threshold (can detect more subtle changes)
        if len(bg_values) > 0:
            bg_mean = np.mean(bg_values)
            bg_std = np.std(bg_values)

            # Scale threshold inversely with SNR
            # At SNR=10: threshold = bg_mean + 4*bg_std
            # At SNR=1000: threshold = bg_mean + 2*bg_std
            std_multiplier = 4.0 - 2.0 * snr_factor
            env_threshold = max(0.03, min(0.3, bg_mean + std_multiplier * bg_std))
        else:
            # Fallback threshold based purely on SNR
            env_threshold = 0.10 - 0.07 * snr_factor  # 0.10 at low SNR, 0.03 at high SNR

        # Find envelope window - start from merger and work backwards
        # For higher SNR, require fewer consecutive points below threshold
        required_below = max(1, int(5 - 3 * snr_factor))  # 5 at low SNR, 2 at high SNR

        env_start_idx = merger_idx
        consecutive_below = 0

        while env_start_idx > 0 and consecutive_below < required_below:
            if smooth_envelope[env_start_idx] <= env_threshold:
                consecutive_below += 1
            else:
                consecutive_below = 0
            env_start_idx -= 1

        # Convert to time
        env_window_before = actual_merger_time - times[env_start_idx + consecutive_below]

        # Sanity checks without relying on masses
        # Cap window based on SNR
        max_expected_window = 0.05 + 0.35 * snr_factor  # 50ms at SNR=10, 400ms at SNR=1000
        if env_window_before > max_expected_window:
            print(f"     Envelope detection found suspiciously large window: {env_window_before:.4f}s")
            print(f"     Capping to SNR-appropriate maximum: {max_expected_window:.4f}s")
            env_window_before = max_expected_window
        else:
            print(f"     Envelope detection found window: {env_window_before:.4f}s before merger (threshold: {env_threshold:.4f})")
    except Exception as e:
        print(f"     Envelope detection failed: {e}. Using fallback method.")
        env_window_before = snr_window_before * 0.8

    # --------------------------------------------------------------------------------
    # METHOD 3: SPECTRAL VARIANCE (SNR-OPTIMIZED)
    # --------------------------------------------------------------------------------
    print("     Method 3: Analyzing spectral variance with SNR-adaptive parameters...")

    try:
        # Window length for spectral analysis - adapt based on SNR
        # Higher SNR allows shorter windows (better time resolution)
        # Lower SNR requires longer windows (better frequency resolution)
        spec_window_seconds = 0.015 - 0.010 * snr_factor  # 15ms at low SNR, 5ms at high SNR
        window_len = int(white_data_bp.sample_rate * spec_window_seconds)

        # Step size also adapts with SNR (more overlap at high SNR)
        step_factor = 0.5 - 0.3 * snr_factor  # 50% step at low SNR, 20% step at high SNR
        step = max(1, int(window_len * step_factor))

        # Calculate spectral variance
        spec_var = []
        spec_times = []

        for i in range(0, len(values) - window_len, step):
            if i + window_len <= len(values):
                segment = values[i:i+window_len]
                segment_fft = np.abs(np.fft.rfft(segment))**2

                # Use adaptive frequency range based on SNR
                # For low SNR, focus on mid-range frequencies where GW signal is strongest
                # For high SNR, include more of the spectrum
                if snr < 20:
                    # Focus on 50-150 Hz band for low SNR
                    freq_idx_min = int(50 * window_len / white_data_bp.sample_rate)
                    freq_idx_max = int(150 * window_len / white_data_bp.sample_rate)
                    freq_band = segment_fft[freq_idx_min:freq_idx_max+1]
                    var_metric = np.std(freq_band) if len(freq_band) > 0 else 0
                else:
                    # Use wider band for higher SNR
                    freq_idx_min = int(30 * window_len / white_data_bp.sample_rate)
                    var_metric = np.std(segment_fft[freq_idx_min:])

                spec_var.append(var_metric)
                spec_times.append(times[i + window_len//2])

        # Normalize spectral variance
        if spec_var and max(spec_var) > 0:
            spec_var = np.array(spec_var) / np.max(spec_var)

            # Get background from first portion
            noise_idx = int(0.2 * len(spec_var))
            bg_values = spec_var[:noise_idx]

            # Robust median-based statistics
            bg_median = np.median(bg_values)
            bg_mad = np.median(np.abs(bg_values - bg_median)) * 1.4826  # Median absolute deviation

            # SNR-adaptive threshold
            # Lower threshold for higher SNR
            std_multiplier = 4.0 - 2.5 * snr_factor
            spec_threshold = max(0.05, min(0.3, bg_median + std_multiplier * bg_mad))

            # Apply smoothing (more smoothing for lower SNR)
            window_size = max(3, int(7 - 4 * snr_factor))
            if window_size % 2 == 0:
                window_size += 1  # Ensure odd window size
            smoothed_spec = np.convolve(spec_var, np.ones(window_size)/window_size, mode='same')

            # Find first point exceeding threshold
            spec_start_idx = noise_idx
            while spec_start_idx < len(smoothed_spec) and smoothed_spec[spec_start_idx] < spec_threshold:
                spec_start_idx += 1

            if spec_start_idx < len(smoothed_spec):
                # Convert to time
                spec_window_before = actual_merger_time - spec_times[spec_start_idx]

                # Cap window based on SNR (same as envelope method)
                max_expected_window = 0.05 + 0.35 * snr_factor
                if spec_window_before > max_expected_window:
                    print(f"     Spectral variance found suspiciously large window: {spec_window_before:.4f}s")
                    print(f"     Capping to SNR-appropriate maximum: {max_expected_window:.4f}s")
                    spec_window_before = max_expected_window
                else:
                    print(f"     Spectral variance found window: {spec_window_before:.4f}s before merger (threshold: {spec_threshold:.4f})")
            else:
                spec_window_before = snr_window_before * 0.6
                print(f"     Spectral variance method failed to detect signal onset. Using {spec_window_before:.4f}s (60% of SNR window)")
        else:
            spec_window_before = snr_window_before * 0.6
            print(f"     Spectral variance method failed - insufficient data. Using {spec_window_before:.4f}s")
    except Exception as e:
        print(f"     Spectral analysis failed: {e}. Using fallback method.")
        spec_window_before = snr_window_before * 0.6

    # --------------------------------------------------------------------------------
    # COLLECT WINDOW ESTIMATES AND MAKE DECISION
    # --------------------------------------------------------------------------------
    window_estimates = [
        ("SNR-based", snr_window_before),
        ("Envelope", env_window_before),
        ("Spectral", spec_window_before)
    ]

    # Calculate optimal window using weighted average - weights depend on SNR
    if snr > 100:  # High SNR - data-driven methods more reliable
        print("     High SNR detected: Weighting envelope detection more heavily")
        weights = [0.15, 0.55, 0.3]  # [SNR-based, Envelope, Spectral]
    elif snr > 30:  # Medium SNR - balanced approach
        print("     Medium SNR detected: Using balanced weights")
        weights = [0.25, 0.45, 0.3]
    else:  # Low SNR - rely more on SNR-based scaling
        print("     Low SNR detected: Weighting SNR-based method more heavily")
        weights = [0.2, 0.6, 0.2]

    # Calculate weighted average
    window_before = sum(w * est[1] for w, est in zip(weights, window_estimates))
    print(f"     Using weighted average: {weights[0]:.1f}*SNR + {weights[1]:.1f}*envelope + {weights[2]:.1f}*spectral")

    # Ensure window is reasonable (limits based on SNR)
    min_window_before = 0.01 + 0.04 * snr_factor  # 10ms at low SNR, 50ms at high SNR
    max_window_before = 0.10 + 0.30 * snr_factor  # 100ms at low SNR, 400ms at high SNR
    window_before = max(min_window_before, min(window_before, max_window_before))

    # After-merger window (simpler, just based on SNR)
    window_after = snr_window_after  # Already calculated from SNR

    print(f"     Final window: {window_before:.4f}s before and {window_after:.4f}s after merger")

    # --------------------------------------------------------------------------------
    # VISUALIZATION
    # --------------------------------------------------------------------------------
    plt.figure(figsize=(12, 9))

    # Plot 1: Signal with all window estimates
    plt.subplot(2, 1, 1)
    plt.plot(times, values, 'b-', label='Whitened Signal')
    plt.axvline(actual_merger_time, color='g', linestyle='-', linewidth=2, label=f'Merger Time')

    colors = ['r', 'm', 'y']
    for i, (name, window) in enumerate(window_estimates):
        plt.axvline(actual_merger_time - window, color=colors[i], linestyle='--',
                   label=f'{name} Window: {window:.4f}s')

    plt.axvline(actual_merger_time - window_before, color='k', linestyle='-', linewidth=2,
               label=f'FINAL Window: {window_before:.4f}s')

    plt.title('Signal with Window Estimates')
    plt.xlabel('Time (s)')
    plt.ylabel('Whitened Strain')
    plt.grid(True)
    plt.legend(loc='upper right')

    # Plot 2: Zoomed-in view
    plt.subplot(2, 1, 2)
    plt.plot(times, values, 'b-', label='Whitened Signal')
    plt.axvline(actual_merger_time, color='g', linestyle='-', linewidth=2, label=f'Merger Time')

    for i, (name, window) in enumerate(window_estimates):
        plt.axvline(actual_merger_time - window, color=colors[i], linestyle='--',
                   label=f'{name} Window: {window:.4f}s')

    plt.axvline(actual_merger_time - window_before, color='k', linestyle='-', linewidth=2,
               label=f'FINAL Window: {window_before:.4f}s')

    plt.title('Zoomed Window View')
    plt.xlabel('Time (s)')
    plt.ylabel('Whitened Strain')
    plt.grid(True)
    plt.xlim(actual_merger_time - max(window_before * 1.5, 0.3), actual_merger_time + 0.05)
    plt.legend(loc='upper right')

    plt.tight_layout()
    plt.show()

    # # Add a watermark timestamp to the plot
    # current_datetime = datetime.datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S")
    # current_user = "1Aadi0"
    # plt.figtext(0.01, 0.01, f"{current_datetime} | {current_user}",
    #             fontsize=8, alpha=0.7)

    return window_before, window_after

# -----------------------------------------------------------------------------
# SECTION 1.7: SNR-ONLY ZERO CROSSING ANALYSIS (NEW)
# -----------------------------------------------------------------------------

def analyze_zero_crossings_with_phase_separation(segment_times, segment_values, actual_merger_time, params):
    """SNR-only wrapper function for backward compatibility"""
    # Extract only SNR from params
    snr = params.get('snr', 30)  # Default to 30 if not found
    return analyze_zero_crossings_snr_only(segment_times, segment_values, actual_merger_time, snr)

def adaptive_zero_crossing_detection(times, values, snr):
    """
    Enhanced zero crossing detection with SNR-adaptive smoothing and multi-layer validation
    """
    from scipy.signal import savgol_filter, medfilt

    # Step 1: SNR-adaptive preprocessing
    if snr < 30:
        print("     Applying heavy smoothing for low SNR signal")
        # Heavy smoothing for very low SNR
        window_size = min(15, len(values)//8)
        median_size = min(5, len(values)//10)

        # Apply median filter first to remove impulse noise
        if median_size >= 3:
            smoothed_values = medfilt(values, kernel_size=median_size)
        else:
            smoothed_values = values

        # Then apply Savitzky-Golay smoothing
        if window_size >= 5 and len(values) >= window_size:
            if window_size % 2 == 0:
                window_size += 1
            try:
                smoothed_values = savgol_filter(smoothed_values, window_size, 2)
            except:
                smoothed_values = values

    elif snr < 100:
        print("     Applying moderate smoothing for medium SNR signal")
        # Moderate smoothing for medium SNR
        window_size = min(7, len(values)//10)

        if window_size >= 5 and len(values) >= window_size:
            if window_size % 2 == 0:
                window_size += 1
            try:
                smoothed_values = savgol_filter(values, window_size, 2)
            except:
                smoothed_values = values
        else:
            smoothed_values = values

    else:
        print("     Minimal smoothing for high SNR signal")
        # Minimal smoothing for high SNR - preserve signal fidelity
        window_size = 3
        if len(values) >= 3:
            try:
                smoothed_values = savgol_filter(values, window_size, 1)
            except:
                smoothed_values = values
        else:
            smoothed_values = values

    # Step 2: Enhanced envelope calculation for quality assessment
    try:
        from scipy.signal import hilbert
        analytic_signal = hilbert(smoothed_values)
        envelope = np.abs(analytic_signal)

        # Normalize envelope
        if np.max(envelope) > 0:
            envelope = envelope / np.max(envelope)
        else:
            envelope = np.ones_like(smoothed_values)

        # Apply additional envelope smoothing for low SNR
        if snr < 50:
            env_window = min(7, len(envelope)//8)
            if env_window >= 3 and env_window % 2 == 1:
                try:
                    envelope = savgol_filter(envelope, env_window, 1)
                except:
                    pass

    except Exception as e:
        print(f"     Envelope calculation failed: {e}, using unity envelope")
        envelope = np.ones_like(smoothed_values)

    # Step 3: Multi-threshold zero crossing detection
    print("     Applying multi-threshold zero crossing detection")

    zero_crossings = []
    crossing_types = []
    crossing_qualities = []
    crossing_confidences = []

    # Calculate adaptive thresholds based on signal statistics
    signal_std = np.std(smoothed_values)
    signal_mean = np.mean(smoothed_values)

        # FIXED (LESS AGGRESSIVE)
    # REPLACE WITH:
    if snr < 30:
        threshold_factors = [0.0, 0.2, -0.2]
        quality_threshold = 0.05
    elif snr < 100:
        threshold_factors = [0.0, 0.15, -0.15]
        quality_threshold = 0.1
    elif snr < 200:
        threshold_factors = [0.0, 0.08, -0.08]
        quality_threshold = 0.2
    else:
        # Higher SNR requires stricter thresholds to avoid false positives
        threshold_factors = [0.0, 0.05, -0.05]  # Stricter for very high SNR
        quality_threshold = 0.3  # Higher threshold = stricter filtering


    # Apply multiple threshold levels
    all_crossings = []

    for threshold_factor in threshold_factors:
        threshold = threshold_factor * signal_std

        for i in range(1, len(smoothed_values)):
            crossing_type = None

            # Rising zero crossing with threshold
            if (smoothed_values[i-1] - threshold) < 0 and (smoothed_values[i] - threshold) >= 0:
                crossing_type = 1
            # Falling zero crossing with threshold
            elif (smoothed_values[i-1] - threshold) >= 0 and (smoothed_values[i] - threshold) < 0:
                crossing_type = -1

            if crossing_type is not None:
                if snr > 100:
                    # Get a few points around the crossing
                    window_start = max(0, i-3)
                    window_end = min(len(smoothed_values), i+4)
                    window_values = smoothed_values[window_start:window_end]

                    # Check if signal truly changes sign
                    signs_change = False
                    for j in range(len(window_values)-1):
                        if window_values[j] * window_values[j+1] <= 0:  # Zero or sign change
                            signs_change = True
                            break

                    if not signs_change:
                        crossing_type = None  # Invalidate the crossing
                        continue  # Skip
                # Enhanced interpolation for precise crossing time
                y1, y2 = smoothed_values[i-1] - threshold, smoothed_values[i] - threshold

                # FIXED: Proper line break and indentation
                if abs(y2 - y1) > 1e-12:  # Avoid division by zero
                    t_cross = times[i-1] + (times[i] - times[i-1]) * abs(y1) / abs(y2 - y1)
                else:
                    t_cross = times[i]

                # Enhanced quality scoring with multiple factors
                local_snr = envelope[i]
                steepness = abs(smoothed_values[i] - smoothed_values[i-1]) / (times[i] - times[i-1])

                # Signal clarity factor (how far from noise level)
                signal_strength = abs(smoothed_values[i]) / max(signal_std, 1e-12)

                # Combined quality score
                quality = local_snr * steepness * (1 + signal_strength) * (1 + snr/100)

                # Confidence score based on local signal consistency
                window_start = max(0, i-2)
                window_end = min(len(smoothed_values), i+3)
                local_window = smoothed_values[window_start:window_end]
                local_std = np.std(local_window)
                local_trend = abs(np.mean(local_window[:len(local_window)//2]) -
                                np.mean(local_window[len(local_window)//2:]))

                confidence = min(1.0, local_trend / max(local_std, 1e-12))

                all_crossings.append({
                    'time': t_cross,
                    'type': crossing_type,
                    'quality': quality,
                    'confidence': confidence,
                    'threshold': threshold_factor,
                    'steepness': steepness,
                    'envelope': local_snr
                })

    # Step 4: Consolidate and filter crossings
    print(f"     Found {len(all_crossings)} raw crossings from multi-threshold detection")

    # Remove duplicates (crossings too close together)
    min_separation = 1.0 / (2 * (600 + 200 * min(1.0, snr/100)))  # Adaptive based on max expected freq

    # Sort by time
    all_crossings.sort(key=lambda x: x['time'])

    # Remove nearby duplicates, keeping highest quality
    filtered_crossings = []
    for crossing in all_crossings:
        # Check if too close to existing crossings
        too_close = False
        for existing in filtered_crossings:
            if abs(crossing['time'] - existing['time']) < min_separation:
                # If new crossing is higher quality, replace existing
                if crossing['quality'] > existing['quality']:
                    filtered_crossings.remove(existing)
                    break
                else:
                    too_close = True
                    break

        if not too_close:
            filtered_crossings.append(crossing)

    # Step 5: Final quality filtering with fallback
    if filtered_crossings:
        # Calculate adaptive quality threshold
        qualities = [c['quality'] for c in filtered_crossings]
        quality_median = np.median(qualities)
        quality_mad = np.median(np.abs(np.array(qualities) - quality_median))

        # More permissive threshold for low SNR
        # FIXED (MORE PERMISSIVE)
        # FIXED - Much more permissive for high SNR
        if snr > 200:
            quality_cutoff = quality_median - 8 * quality_mad  # Very permissive for high SNR
        elif snr > 100:
            quality_cutoff = quality_median - 6 * quality_mad  # Moderately permissive
        else:
            quality_cutoff = quality_median - 4 * quality_mad  # Still reasonable for low SNR

        print(f"     Using quality cutoff: {quality_cutoff:.3e} (SNR={snr})")



        # Apply quality filter
        high_quality_crossings = [c for c in filtered_crossings if c['quality'] >= quality_cutoff]

        print(f"     After quality filtering: {len(high_quality_crossings)}/{len(filtered_crossings)} crossings")

        # Extract final results
        for crossing in high_quality_crossings:
            zero_crossings.append(crossing['time'])
            crossing_types.append(crossing['type'])
            crossing_qualities.append(crossing['quality'])
            crossing_confidences.append(crossing['confidence'])

        # FALLBACK: If no high-quality crossings, use lower threshold
        if not zero_crossings and len(filtered_crossings) > 0:
            print("     No high-quality crossings found, applying relaxed threshold...")
            relaxed_cutoff = quality_median - 5 * quality_mad  # Much more permissive

            relaxed_crossings = [c for c in filtered_crossings if c['quality'] >= relaxed_cutoff]

            for crossing in relaxed_crossings:
                zero_crossings.append(crossing['time'])
                crossing_types.append(crossing['type'])
                crossing_qualities.append(crossing['quality'])
                crossing_confidences.append(crossing['confidence'])

            print(f"     Relaxed threshold yielded: {len(zero_crossings)} crossings")

    # ULTIMATE FALLBACK: Simple zero crossing if enhanced method fails
    if not zero_crossings:
        print("     Enhanced detection failed, falling back to simple zero crossing...")

        # FIXED: Proper line break and indentation
        # Simple zero crossing detection as backup
        for i in range(1, len(smoothed_values)):
            if (smoothed_values[i-1] < 0 and smoothed_values[i] >= 0) or \
               (smoothed_values[i-1] >= 0 and smoothed_values[i] < 0):

                # Simple interpolation
                if abs(smoothed_values[i] - smoothed_values[i-1]) > 1e-12:
                    t_cross = times[i-1] + (times[i] - times[i-1]) * abs(smoothed_values[i-1]) / abs(smoothed_values[i] - smoothed_values[i-1])
                else:
                    t_cross = times[i]

                crossing_type = 1 if smoothed_values[i-1] < 0 else -1

                zero_crossings.append(t_cross)
                crossing_types.append(crossing_type)
                crossing_qualities.append(0.5)  # Default quality
                crossing_confidences.append(0.5)  # Default confidence

        print(f"     Simple fallback found: {len(zero_crossings)} crossings")

    # Final validation
    if zero_crossings:
        print(f"     Final result: {len(zero_crossings)} zero crossings returned")
    else:
        print("     WARNING: No zero crossings found even with fallback!")

    return zero_crossings, crossing_types, crossing_qualities, crossing_confidences

def enhanced_frequency_estimation(zero_crossings, crossing_types, snr, actual_merger_time):
    """
    Enhanced frequency estimation with outlier rejection and physics constraints
    """
    from scipy import stats
    from scipy.optimize import curve_fit
    import warnings
    warnings.filterwarnings('ignore')

    print(f"     Applying enhanced frequency estimation (SNR={snr})")

    # Step 1: Calculate basic frequencies
    basic_freqs = []
    basic_times = []

    for i in range(len(zero_crossings) - 1):
        # Skip non-alternating pairs for better accuracy
        if i < len(crossing_types) - 1 and crossing_types[i] * crossing_types[i+1] > 0:
            continue

        delta_t = zero_crossings[i+1] - zero_crossings[i]
        if delta_t > 0:
            freq = 1.0 / (2 * delta_t)
            mid_time = (zero_crossings[i] + zero_crossings[i+1]) / 2
            basic_freqs.append(freq)
            basic_times.append(mid_time)

    if len(basic_freqs) < 3:
        print(f"     Too few frequency points ({len(basic_freqs)}), returning basic calculation")
        return basic_times, basic_freqs

    print(f"     Calculated {len(basic_freqs)} basic frequency points")

    # Step 2: Outlier detection using z-score
    z_scores = np.abs(stats.zscore(basic_freqs))

        # REPLACE WITH:
    if snr < 30:
        outlier_threshold = 4.0  # More permissive for noisy data
    elif snr < 100:
        outlier_threshold = 3.5  # Moderately permissive
    elif snr < 250:
        outlier_threshold = 2.5  # More strict
    else:
        outlier_threshold = 2.0  # Much stricter for very high SNR


    valid_mask = z_scores < outlier_threshold

    clean_times = np.array(basic_times)[valid_mask]
    clean_freqs = np.array(basic_freqs)[valid_mask]

    outliers_removed = len(basic_freqs) - len(clean_freqs)
    if outliers_removed > 0:
        print(f"     Removed {outliers_removed} outlier(s) using z-score analysis")

    if len(clean_freqs) < 3:
        print(f"     Too many outliers removed, using original data")
        return basic_times, basic_freqs

    # Step 3: Enforce monotonic frequency increase (physical constraint)
    print("     Enforcing monotonic frequency evolution")
    monotonic_times = [clean_times[0]]
    monotonic_freqs = [clean_freqs[0]]

    for i in range(1, len(clean_times)):
        # Allow small decreases (2%) for noise tolerance
        if clean_freqs[i] >= monotonic_freqs[-1] * 0.98:
            monotonic_times.append(clean_times[i])
            monotonic_freqs.append(clean_freqs[i])
        else:
            # Gentle correction: use slight increase over previous
            corrected_freq = monotonic_freqs[-1] * 1.01
            monotonic_times.append(clean_times[i])
            monotonic_freqs.append(corrected_freq)
            print(f"     Corrected non-monotonic frequency at t={clean_times[i]:.4f}s")

    # Step 4: Post-Newtonian frequency evolution fitting (for SNR > 50)
    if snr > 50 and len(monotonic_freqs) > 5:
        print("     Applying post-Newtonian frequency evolution fitting")

        try:
            # Theoretical model: f(t) ∝ (τ-t)^(-3/8) for inspiral
            def pn_freq_model(t, f0, tau, gamma):
                # f0: frequency scale, tau: coalescence time, gamma: evolution rate
                time_to_merger = np.maximum(tau - t, 0.001)  # Avoid division by zero
                return f0 * np.power(time_to_merger, gamma)

            # Initial parameter guess
            f0_guess = monotonic_freqs[0]
            tau_guess = actual_merger_time + 0.01  # Slightly after merger
            gamma_guess = -3/8  # Theoretical PN value
            # ADD these constraints:
            # Constrain gamma based on SNR
            if snr > 200:
                gamma_bounds = (-0.385, -0.365)  # Very tight around theoretical -3/8 = -0.375
            elif snr > 100:
                gamma_bounds = (-0.395, -0.355)  # Tight constraints
            else:
                gamma_bounds = (-0.5, -0.25)     # Loose constraints for low SNR


            # Fit the model
            popt, pcov = curve_fit(
                pn_freq_model,
                monotonic_times,
                monotonic_freqs,
                p0=[f0_guess, tau_guess, gamma_guess],
                bounds=([f0_guess*0.1, actual_merger_time-0.1, gamma_bounds[0]],
       [f0_guess*10, actual_merger_time+0.1, gamma_bounds[1]]),
                maxfev=1000
            )

            theoretical_freqs = pn_freq_model(np.array(monotonic_times), *popt)

            # SNR-adaptive blending
            if snr > 100:
                blend_factor = 0.8  # Trust observations more for high SNR
            elif snr > 75:
                blend_factor = 0.7  # Balanced blending
            else:
                blend_factor = 0.6  # Trust theory more for medium SNR

            # Blend observed and theoretical frequencies
            enhanced_freqs = (blend_factor * np.array(monotonic_freqs) +
                            (1-blend_factor) * theoretical_freqs)

            print(f"     PN fitting successful, blending with factor {blend_factor:.1f}")
            print(f"     Fitted parameters: f0={popt[0]:.1f}Hz, τ={popt[1]:.4f}s, γ={popt[2]:.3f}")

            return monotonic_times, enhanced_freqs.tolist()

        except Exception as e:
            print(f"     PN fitting failed: {str(e)[:50]}... using monotonic frequencies")

    # Step 5: Additional smoothing for low SNR
    if snr < 50 and len(monotonic_freqs) > 5:
        print("     Applying additional smoothing for low SNR")
        from scipy.signal import savgol_filter

        # Apply gentle smoothing
        window_length = min(7, len(monotonic_freqs)//2*2+1)  # Ensure odd
        if window_length >= 3:
            try:
                smoothed_freqs = savgol_filter(monotonic_freqs, window_length, 2)
                monotonic_freqs = smoothed_freqs.tolist()
                print(f"     Applied Savitzky-Golay smoothing (window={window_length})")
            except:
                pass

    print(f"     Enhanced frequency estimation complete: {len(monotonic_times)} points")
    return monotonic_times, monotonic_freqs



def analyze_zero_crossings_snr_only(segment_times, segment_values, actual_merger_time, snr):
    """
    IMPROVED Zero-crossing analysis with robust filtering based ONLY on SNR.
    Simplified approach with fewer, better-justified filtering stages.

    Parameters:
        segment_times: Time array for the signal segment
        segment_values: Amplitude values for the signal segment
        actual_merger_time: Time of the merger
        snr: Signal-to-noise ratio

    Returns:
        Dictionary containing zero crossing analysis results
    """
    print(f"     Starting SNR-only zero crossing analysis (SNR={snr})...")

    # Input validation
    if len(segment_times) != len(segment_values):
        raise ValueError(f"Array length mismatch: times={len(segment_times)}, values={len(segment_values)}")

    if len(segment_times) < 10:
        raise ValueError("Insufficient data points for zero crossing analysis")

    # Convert to numpy arrays for better performance
    times = np.array(segment_times)
    values = np.array(segment_values)

    # Calculate SNR scaling factor once
    snr_factor = min(1.0, np.log10(max(10, snr)) / np.log10(1000))
    print(f"     SNR scaling factor: {snr_factor:.3f}")

    # =========================================================================
    # STEP 1: DETECT ALL ZERO CROSSINGS WITH QUALITY SCORES
    # =========================================================================

    zero_crossings = []
    crossing_types = []  # 1 for rising, -1 for falling
    crossing_qualities = []

    # Calculate signal envelope for quality assessment
    try:
        from scipy.signal import hilbert
        envelope = np.abs(hilbert(values))
        envelope = envelope / np.max(envelope) if np.max(envelope) > 0 else np.ones_like(envelope)
    except Exception:
        envelope = np.ones_like(values)

    # Replace the simple detection loop with the enhanced system:
    print("     Applying enhanced adaptive zero-crossing detection...")
    enhanced_crossings, enhanced_types, enhanced_qualities, enhanced_confidences = adaptive_zero_crossing_detection(
        times, values, snr
    )

    # Validate the results
    if enhanced_crossings and len(enhanced_crossings) > 0:
        print(f"     Enhanced detection successful: {len(enhanced_crossings)} crossings")
        zero_crossings = enhanced_crossings
        crossing_types = enhanced_types
        crossing_qualities = enhanced_qualities
        crossing_confidences = enhanced_confidences
    else:
        print("     Enhanced detection failed, no crossings found!")
        # Could add additional fallback here if needed
        zero_crossings = []
        crossing_types = []
        crossing_qualities = []
        crossing_confidences = []

    print(f"     Final zero crossings count: {len(zero_crossings)}")

    if len(zero_crossings) < 4:
        print("     ERROR: Insufficient zero crossings for analysis")
        return _create_empty_result()

    # =========================================================================
    # STEP 2: SIMPLIFIED FILTERING - ONLY ESSENTIAL FILTERS
    # =========================================================================

    # Filter 1: Quality threshold (SNR-adaptive)
        # FIXED - SNR-adaptive quality threshold
    if snr > 200:
        quality_threshold = np.percentile(crossing_qualities, 10)  # Keep 90% of crossings
    elif snr > 100:
        quality_threshold = np.percentile(crossing_qualities, 20)  # Keep 80% of crossings
    else:
        quality_threshold = np.percentile(crossing_qualities, 30)  # Keep 70% of crossings

    quality_mask = np.array(crossing_qualities) >= quality_threshold
    print(f"     Quality threshold: {quality_threshold:.3e} (keeping {np.sum(quality_mask)}/{len(crossing_qualities)} crossings)")
    # More permissive

    filtered_crossings = []
    filtered_types = []
    filtered_qualities = []

    for i, quality in enumerate(crossing_qualities):
        if quality >= quality_threshold:
            filtered_crossings.append(zero_crossings[i])
            filtered_types.append(crossing_types[i])
            filtered_qualities.append(quality)

    print(f"     After quality filter: {len(filtered_crossings)} crossings")

    # Filter 2: Physics-based frequency limits
    f_min = 15.0  # Fixed lower bound based on LIGO sensitivity
    f_max = 300 + 200 * snr_factor  # Adaptive upper bound: 300-500 Hz

        # FIXED - SNR-adaptive spacing that's much more permissive
    if snr > 200:
        min_spacing = 1.0 / (3 * f_max)  # Allow closer crossings for high SNR
    elif snr > 100:
        min_spacing = 1.0 / (2.5 * f_max)  # Moderately permissive
    else:
        min_spacing = 1.0 / (2 * f_max)    # Original for low SNR

    max_spacing = 1.0 / (2 * f_min)  # Keep this unchanged

    print(f"     SNR-adaptive spacing: min={min_spacing*1000:.2f}ms, max={max_spacing*1000:.1f}ms")


    spaced_crossings = []
    spaced_types = []
    spaced_qualities = []

    if filtered_crossings:
        spaced_crossings.append(filtered_crossings[0])
        spaced_types.append(filtered_types[0])
        spaced_qualities.append(filtered_qualities[0])

        for i in range(1, len(filtered_crossings)):
            spacing = filtered_crossings[i] - spaced_crossings[-1]
            if min_spacing <= spacing <= max_spacing:
                spaced_crossings.append(filtered_crossings[i])
                spaced_types.append(filtered_types[i])
                spaced_qualities.append(filtered_qualities[i])

    print(f"     After spacing filter: {len(spaced_crossings)} crossings")

    # Filter 3: Relaxed alternation (only for high SNR)
    if snr > 50 and len(spaced_crossings) > 2:
        alternating_crossings = [spaced_crossings[0]]
        alternating_types = [spaced_types[0]]
        alternating_qualities = [spaced_qualities[0]]

        expected_type = -spaced_types[0]
        consecutive_same = 0

        for i in range(1, len(spaced_crossings)):
            if spaced_types[i] == expected_type:
                # Perfect alternation
                alternating_crossings.append(spaced_crossings[i])
                alternating_types.append(spaced_types[i])
                alternating_qualities.append(spaced_qualities[i])
                expected_type = -expected_type
                consecutive_same = 0
            elif consecutive_same < 1:  # Allow one consecutive same-type crossing
                alternating_crossings.append(spaced_crossings[i])
                alternating_types.append(spaced_types[i])
                alternating_qualities.append(spaced_qualities[i])
                consecutive_same += 1

        zero_crossings = alternating_crossings
        crossing_types = alternating_types
        crossing_qualities = alternating_qualities
        print(f"     After relaxed alternation filter: {len(zero_crossings)} crossings")
    else:
        zero_crossings = spaced_crossings
        crossing_types = spaced_types
        crossing_qualities = spaced_qualities

    # =========================================================================
    # STEP 3: CALCULATE FREQUENCIES AND PHASE CLASSIFICATION
    # =========================================================================

        # Define ringdown start with SNR-adaptive buffer
    ringdown_buffer = 0.0005 + 0.0015 * snr_factor
    ringdown_start = actual_merger_time + ringdown_buffer

    # Calculate enhanced frequencies FIRST, then classify phases
    print("     Calculating enhanced frequencies from zero crossings...")
    enhanced_times, enhanced_freqs = enhanced_frequency_estimation(
        zero_crossings, crossing_types, snr, actual_merger_time
    )

    # Use the enhanced results
    frequencies = enhanced_freqs
    crossing_times = enhanced_times

    print(f"     Enhanced frequency range: {min(frequencies):.1f} - {max(frequencies):.1f} Hz")

    # NOW classify phases using the calculated frequencies and times
    crossing_phase = []
    crossing_valid = []

    for i, (time, freq) in enumerate(zip(crossing_times, frequencies)):
        if time < actual_merger_time and freq < 250:  # Before merger and reasonable frequency
            crossing_phase.append('inspiral')
            crossing_valid.append(True)
        else:
            crossing_phase.append('ringdown')
            crossing_valid.append(True)

    print(f"     Phase classification: {crossing_phase.count('inspiral')} inspiral, {crossing_phase.count('ringdown')} ringdown")


    # # Replace the basic frequency calculation with enhanced estimation:
    # print("     Calculating enhanced frequencies from zero crossings...")
    # enhanced_times, enhanced_freqs = enhanced_frequency_estimation(
    #     zero_crossings, crossing_types, snr, actual_merger_time
    # )

    # # Use the enhanced results
    # frequencies = enhanced_freqs
    # crossing_times = enhanced_times

    # print(f"     Enhanced frequency range: {min(frequencies):.1f} - {max(frequencies):.1f} Hz")

    if len(frequencies) < 3:
        print("     WARNING: Too few frequency points for reliable analysis")
        # Don't fail completely - return what we have

    # =========================================================================
    # STEP 4: CALCULATE DERIVED QUANTITIES
    # =========================================================================

    # Find merger frequency
    merger_freq = None
    if crossing_times:
        merger_idx = np.argmin(np.abs(np.array(crossing_times) - actual_merger_time))
        merger_freq = frequencies[merger_idx]
    else:
        merger_freq = 150 + 100 * snr_factor  # Fallback estimate

    # Separate inspiral and ringdown
    inspiral_indices = [i for i, phase in enumerate(crossing_phase) if phase == 'inspiral']
    inspiral_freqs = [frequencies[i] for i in inspiral_indices]
    inspiral_times = [crossing_times[i] for i in inspiral_indices]
    inspiral_qualities = [crossing_qualities[i] for i in inspiral_indices if i < len(crossing_qualities)]

    # Statistics
    total_crossings = len(zero_crossings)
    rising_count = crossing_types.count(1)
    falling_count = crossing_types.count(-1)

    print(f"     Final results: {total_crossings} crossings ({rising_count} rising, {falling_count} falling)")
    print(f"     Inspiral points: {len(inspiral_freqs)}, Ringdown points: {len(frequencies) - len(inspiral_freqs)}")
    print(f"     Merger frequency: {merger_freq:.1f} Hz")

    # =========================================================================
    # RETURN RESULTS
    # =========================================================================

    return {
        'zero_crossings': zero_crossings,
        'crossing_types': crossing_types,
        'frequencies': frequencies,
        'crossing_times': crossing_times,
        'crossing_phase': crossing_phase,
        'crossing_qualities': crossing_qualities[:len(frequencies)],  # Ensure same length
        'merger_freq': merger_freq,
        'inspiral_freqs': inspiral_freqs,
        'inspiral_times': inspiral_times,
        'inspiral_qualities': inspiral_qualities,
        'snr_factor': snr_factor,
        'f_min': f_min,
        'f_max': f_max,
        'statistics': {
            'total_crossings': total_crossings,
            'rising_count': rising_count,
            'falling_count': falling_count,
            'valid_frequencies': len(frequencies),
            'retention_rate': len(frequencies) / max(1, len(zero_crossings)) * 100
        }
    }


def _create_empty_result():
    """Helper function to return empty result when analysis fails"""
    return {
        'zero_crossings': [],
        'crossing_types': [],
        'frequencies': [],
        'crossing_times': [],
        'crossing_phase': [],
        'crossing_qualities': [],
        'merger_freq': None,
        'inspiral_freqs': [],
        'inspiral_times': [],
        'inspiral_qualities': [],
        'snr_factor': 0.0,
        'f_min': 15.0,
        'f_max': 300.0,
        'statistics': {
            'total_crossings': 0,
            'rising_count': 0,
            'falling_count': 0,
            'valid_frequencies': 0,
            'retention_rate': 0.0
        }
    }


# Additional helper function for robust error handling
def validate_zero_crossing_inputs(segment_times, segment_values, actual_merger_time, snr):
    """
    Validate inputs to zero crossing analysis with clear error messages
    """
    if not isinstance(segment_times, (list, np.ndarray)):
        raise TypeError("segment_times must be array-like")

    if not isinstance(segment_values, (list, np.ndarray)):
        raise TypeError("segment_values must be array-like")

    if len(segment_times) != len(segment_values):
        raise ValueError(f"Length mismatch: {len(segment_times)} times vs {len(segment_values)} values")

    if len(segment_times) < 10:
        raise ValueError(f"Insufficient data: {len(segment_times)} points (minimum 10 required)")

    if not (0 < snr < 10000):
        raise ValueError(f"Invalid SNR: {snr} (must be between 0 and 10000)")

    time_range = max(segment_times) - min(segment_times)
    if time_range < 0.001:  # Less than 1ms
        raise ValueError(f"Time range too small: {time_range:.6f}s")

    return True


# Usage example with proper error handling:
def analyze_zero_crossings_with_validation(segment_times, segment_values, actual_merger_time, snr):
    """
    Wrapper function that includes validation and error handling
    """
    try:
        # Validate inputs
        validate_zero_crossing_inputs(segment_times, segment_values, actual_merger_time, snr)

        # Perform analysis
        result = analyze_zero_crossings_snr_only(segment_times, segment_values, actual_merger_time, snr)

        # Additional validation of results
        if result['statistics']['retention_rate'] < 10:
            print(f"WARNING: Low retention rate ({result['statistics']['retention_rate']:.1f}%)")

        return result

    except Exception as e:
        print(f"Zero crossing analysis failed: {e}")
        return _create_empty_result()


# def estimate_chirp_mass_robustly(result, actual_merger_time, snr):
#     """
#     Robust chirp mass estimation from frequency evolution data.
#     Uses multiple enhanced techniques to ensure reliable estimates.

#     Parameters:
#         result: Dictionary from analyze_zero_crossings_snr_only
#         actual_merger_time: Time of merger
#         snr: Signal-to-noise ratio

#     Returns:
#         Dictionary with chirp mass estimation results
#     """
#     inspiral_freqs = result.get('inspiral_freqs', [])
#     inspiral_times = result.get('inspiral_times', [])
#     inspiral_qualities = result.get('inspiral_qualities', [])

#     # Convert SNR to scaling factor
#     snr_factor = min(1.0, np.log10(max(10, snr)) / np.log10(1000))

#     if len(inspiral_freqs) < 5:
#         print("     ERROR: Too few inspiral points for reliable chirp mass estimation")
#         return {
#             'chirp_mass': None,
#             'chirp_mass_error': None,
#             'slope': None,
#             'slope_error': None,
#             'r_squared': 0.0
#         }

#     # IMPROVEMENT 1: Apply more permissive frequency cleaning for chirp mass estimation
#     max_expected_freq = 200 + 300 * snr_factor  # 200 Hz low SNR, 500 Hz high SNR
#     max_jump_ratio = 3.0 - 0.7 * snr_factor  # Less strict at all SNRs (3.0 low, 2.3 high)

#     # Initialize clean arrays
#     clean_freqs = []
#     clean_times = []
#     clean_f83 = []
#     clean_qualities = []

#     # Start with highest quality point
#     if inspiral_qualities:
#         start_idx = np.argmax(inspiral_qualities)
#         clean_freqs.append(inspiral_freqs[start_idx])
#         clean_times.append(inspiral_times[start_idx])
#         clean_f83.append(inspiral_freqs[start_idx]**(-8/3))
#         clean_qualities.append(inspiral_qualities[start_idx])

#     # IMPROVEMENT 2: Two-pass cleaning
#     # First pass - build a sequence based on continuity
#     for i in range(len(inspiral_freqs)):
#         # Skip if this is the starting point or implausibly high frequency
#         if (inspiral_qualities and i == start_idx) or inspiral_freqs[i] > max_expected_freq:
#             continue

#         # Try to add this point to the clean sequence
#         valid_addition = True

#         if clean_freqs:
#             # Find closest time point already in clean sequence
#             time_diffs = np.abs(np.array(clean_times) - inspiral_times[i])
#             closest_idx = np.argmin(time_diffs)

#             # Only check frequency jump if points are close enough in time
#             if time_diffs[closest_idx] < 0.03:  # 30ms threshold
#                 # Check for excessive jump
#                 ratio = max(inspiral_freqs[i]/clean_freqs[closest_idx],
#                            clean_freqs[closest_idx]/inspiral_freqs[i])
#                 if ratio > max_jump_ratio:
#                     valid_addition = False

#         if valid_addition:
#             clean_freqs.append(inspiral_freqs[i])
#             clean_times.append(inspiral_times[i])
#             clean_f83.append(inspiral_freqs[i]**(-8/3))
#             clean_qualities.append(inspiral_qualities[i] if inspiral_qualities else 1.0)

#     # IMPROVEMENT 3: Ensure monotonicity in final sequence
#     # Sort by time
#     if clean_times:
#         sorted_indices = np.argsort(clean_times)
#         clean_freqs = [clean_freqs[i] for i in sorted_indices]
#         clean_times = [clean_times[i] for i in sorted_indices]
#         clean_f83 = [clean_f83[i] for i in sorted_indices]
#         clean_qualities = [clean_qualities[i] for i in sorted_indices]

#         # Remove points that create non-monotonicity in frequency
#         monotonic_freqs = [clean_freqs[0]]
#         monotonic_times = [clean_times[0]]
#         monotonic_f83 = [clean_f83[0]]
#         monotonic_qualities = [clean_qualities[0]]

#         for i in range(1, len(clean_times)):
#             # Only add if frequency is higher than the last added point
#             if clean_freqs[i] >= monotonic_freqs[-1] * 0.98:  # Allow tiny drop (2%)
#                 monotonic_freqs.append(clean_freqs[i])
#                 monotonic_times.append(clean_times[i])
#                 monotonic_f83.append(clean_f83[i])
#                 monotonic_qualities.append(clean_qualities[i])

#         clean_freqs = monotonic_freqs
#         clean_times = monotonic_times
#         clean_f83 = monotonic_f83
#         clean_qualities = monotonic_qualities

#     print(f"     Final cleaned points for chirp mass estimation: {len(clean_freqs)}")

#     if len(clean_freqs) < 5:
#         print("     WARNING: Too few clean points remain for reliable chirp mass estimation")
#         if len(clean_freqs) < 3:
#             return {
#                 'chirp_mass': None,
#                 'chirp_mass_error': None,
#                 'slope': None,
#                 'slope_error': None,
#                 'r_squared': 0.0
#             }

#     # Convert to numpy arrays
#     times_array = np.array(clean_times)
#     f83_array = np.array(clean_f83)
#     qualities_array = np.array(clean_qualities)

#     # IMPROVEMENT 4: ENHANCED WEIGHTING STRATEGY
#     # Weight by combination of:
#     # 1. Quality of measurement
#     # 2. Proximity to merger (later points weighted higher)
#     # 3. SNR-dependent overall scheme

#     time_range = max(times_array) - min(times_array)
#     if time_range > 0:
#         # Time-based weight component
#         time_weights = np.exp((times_array - min(times_array)) / time_range * (3 + snr_factor * 2))

#         # Combine with quality weights
#         if np.max(qualities_array) > 0:
#             quality_weights = qualities_array / np.max(qualities_array)
#             weights = time_weights * (0.3 + 0.7 * quality_weights)  # Balance time and quality
#         else:
#             weights = time_weights

#         # Normalize weights
#         weights = weights / np.mean(weights) * len(weights)
#     else:
#         weights = np.ones_like(times_array)

#     # IMPROVEMENT 5: ROBUST REGRESSION WITH HUBER LOSS
#     try:
#         # Try to use Huber regression (robust against outliers)
#         from sklearn.linear_model import HuberRegressor
#         X = times_array - times_array[0]
#         X = X.reshape(-1, 1)  # Reshape for sklearn
#         huber = HuberRegressor(epsilon=1.35, max_iter=100)
#         huber.fit(X, f83_array, sample_weight=weights)

#         slope = huber.coef_[0]
#         intercept = huber.intercept_

#         # Calculate pseudo R^2
#         y_pred = huber.predict(X)
#         residuals = f83_array - y_pred
#         ss_res = np.sum(residuals**2)
#         ss_tot = np.sum((f83_array - np.mean(f83_array))**2)
#         r_squared = max(0, 1 - (ss_res / ss_tot)) if ss_tot > 0 else 0

#         # Estimate slope error using bootstrap
#         slope_errors = []
#         n_bootstrap = min(50, len(times_array) * 5)

#         for _ in range(n_bootstrap):
#             # Resample with replacement
#             indices = np.random.choice(len(times_array), size=len(times_array), replace=True)
#             X_boot = X[indices]
#             y_boot = f83_array[indices]
#             w_boot = weights[indices]

#             try:
#                 huber_boot = HuberRegressor(epsilon=1.35, max_iter=100)
#                 huber_boot.fit(X_boot, y_boot, sample_weight=w_boot)
#                 slope_errors.append(huber_boot.coef_[0])
#             except:
#                 continue

#         if slope_errors:
#             slope_error = np.std(slope_errors)
#         else:
#             slope_error = abs(slope) * 0.5  # Fallback error estimate

#     except (ImportError, Exception):
#         # Fallback to weighted polyfit
#         poly_coeffs = np.polyfit(times_array - times_array[0], f83_array, 1, w=weights)
#         slope = poly_coeffs[0]
#         intercept = poly_coeffs[1]

#         # Calculate R-squared
#         pred_vals = slope * (times_array - times_array[0]) + intercept
#         residuals = f83_array - pred_vals
#         ss_res = np.sum(residuals**2)
#         ss_tot = np.sum((f83_array - np.mean(f83_array))**2)
#         r_squared = max(0, 1 - (ss_res / ss_tot)) if ss_tot > 0 else 0

#         # Estimate slope error
#         n = len(times_array)
#         if n > 2:
#             x_diff = times_array - times_array[0]
#             slope_error = np.sqrt(ss_res / (n - 2)) / np.sqrt(np.sum((x_diff - np.mean(x_diff))**2))
#         else:
#             slope_error = abs(slope) * 0.5  # Fallback error estimate

#     # IMPROVEMENT 6: VERIFY SLOPE IS PHYSICAL (NEGATIVE)
#     # For chirps, f^(-8/3) vs time must have negative slope
#     valid_result = slope < 0

#     if valid_result:
#         # Calculate chirp mass
#         G = 6.67430e-11  # m^3 kg^-1 s^-2
#         c = 299792458.0  # m/s
#         pi = np.pi
#         M_sun = 1.989e30  # kg

#         conversion_factor = (5/256) * (pi)**(-8/3)
#         M_chirp_factor = ((-slope) * conversion_factor)**(3/5) * (c**3/G)
#         M_chirp_solar = M_chirp_factor / M_sun

#         # Calculate error propagation
#         relative_error_slope = slope_error / (-slope)
#         relative_error_chirp_mass = (3/5) * relative_error_slope
#         M_chirp_error = M_chirp_solar * relative_error_chirp_mass

#         # IMPROVEMENT 7: SANITY CHECK ON RESULT
#         # Typical stellar-mass BH chirp masses are 5-50 solar masses
#         if M_chirp_solar < 1 or M_chirp_solar > 150:
#             print(f"     WARNING: Estimated chirp mass {M_chirp_solar:.2f} M☉ is outside typical range")

#             # If R² is poor and estimate is extreme, increase error bars significantly
#             if r_squared < 0.4 and (M_chirp_solar < 3 or M_chirp_solar > 100):
#                 M_chirp_error = max(M_chirp_error, M_chirp_solar * 0.8)
#                 print(f"     Increasing error estimate due to low confidence (R²={r_squared:.2f})")
#     else:
#         print("     ERROR: Positive slope in f^(-8/3) vs time - physically impossible for a chirp")
#         M_chirp_solar = None
#         M_chirp_error = None
#     # Add final quality check before returning results
#     if r_squared < 0.4 and M_chirp_solar is not None:
#         print("     WARNING: Robust estimate has poor fit (R²<0.4). Trying alternative method...")
#         # Use a simple weighted regression on the original inspiral points
#         try:
#             orig_times = np.array(result.get('inspiral_times', []))
#             orig_freqs = np.array(result.get('inspiral_freqs', []))

#             if len(orig_times) >= 3:
#                 # Convert to f^(-8/3) and create weights
#                 f83_array = np.array([f**(-8/3) for f in orig_freqs])

#                 # Ensure we have a valid time range
#                 time_range = max(orig_times) - min(orig_times)
#                 if time_range > 0:
#                     weights = np.exp((orig_times - min(orig_times)) / time_range * 4)
#                 else:
#                     weights = np.ones_like(orig_times)

#                 # Simple weighted regression
#                 poly_coeffs = np.polyfit(orig_times - orig_times[0], f83_array, 1, w=weights)
#                 new_slope = poly_coeffs[0]
#                 new_intercept = poly_coeffs[1]

#                 # Calculate new R-squared
#                 pred_vals = new_slope * (orig_times - orig_times[0]) + new_intercept
#                 residuals = f83_array - pred_vals
#                 ss_res = np.sum(residuals**2)
#                 ss_tot = np.sum((f83_array - np.mean(f83_array))**2)
#                 new_r_squared = max(0, 1 - (ss_res / ss_tot)) if ss_tot > 0 else 0

#                 # Calculate new chirp mass with this slope
#                 if new_slope < 0 and new_r_squared > r_squared:
#                     # This is a better fit - use it
#                     G = 6.67430e-11
#                     c = 299792458.0
#                     pi = np.pi
#                     M_sun = 1.989e30

#                     conversion_factor = (5/256) * (pi)**(-8/3)
#                     new_M_chirp = ((-new_slope) * conversion_factor)**(3/5) * (c**3/G) / M_sun

#                     # Calculate error using bootstrap-inspired approach
#                     if len(orig_times) > 2:
#                         x_diff = orig_times - orig_times[0]
#                         std_err = np.sqrt(ss_res / (len(orig_times) - 2)) / np.sqrt(np.sum((x_diff - np.mean(x_diff))**2))
#                     else:
#                         std_err = abs(new_slope) * 0.3  # Conservative fallback

#                     rel_err = std_err / (-new_slope) if new_slope < 0 else 0.3
#                     new_M_chirp_err = new_M_chirp * (3/5) * rel_err

#                     print(f"     Alternative method gives better fit (R²={new_r_squared:.4f})")
#                     print(f"     Updated estimate: {new_M_chirp:.2f} ± {new_M_chirp_err:.2f} solar masses")

#                     # Return the improved results
#                     return {
#                         'chirp_mass': new_M_chirp,
#                         'chirp_mass_error': new_M_chirp_err,
#                         'slope': new_slope,
#                         'slope_error': std_err,
#                         'r_squared': new_r_squared,
#                         'times_array': orig_times,
#                         'f83_array': f83_array,
#                         'weights': weights
#                     }
#         except Exception as e:
#             print(f"     Alternative method failed: {e}")

#     return {
#         'chirp_mass': M_chirp_solar,
#         'chirp_mass_error': M_chirp_error,
#         'slope': slope,
#         'slope_error': slope_error,
#         'r_squared': r_squared,
#         'times_array': times_array,
#         'f83_array': f83_array,
#         'weights': weights
#     }

def use_theoretical_reference_points(inspiral_times, inspiral_freqs, params, snr):
    """
    For high SNR cases, adds theoretical reference points in early inspiral
    to improve chirp mass estimation.
    """
    if snr < 250 or len(inspiral_times) < 5:
        return inspiral_times, inspiral_freqs  # Only use for high SNR

    # Get binary parameters if available
    mass1 = params.get('mass1', 30)  # Default reasonable values
    mass2 = params.get('mass2', 30)

    # Calculate chirp mass
    chirp_mass = (mass1 * mass2)**(3/5) / (mass1 + mass2)**(1/5)

    # Calculate min and max times
    t_min = min(inspiral_times)
    t_max = max(inspiral_times)
    actual_merger_time = t_max  # Approximate

    # Add reference points in early inspiral (if we're missing early data)
    if min(inspiral_freqs) > 40:  # Missing early inspiral
        print(f"     Adding theoretical reference points for early inspiral")

        # Theoretical time-frequency relation for inspiral:
        # f(t) = (5/256) * (GM_chirp/c^3)^(-5/8) * (t_c - t)^(-3/8)

        # Constants
        G = 6.67430e-11
        c = 299792458.0
        M_sun = 1.989e30
        M_chirp_kg = chirp_mass * M_sun

        # Time offsets from merger (in seconds)
        time_offsets = [0.4, 0.3, 0.25]  # Add points further back in inspiral

        # Add theoretical points
        new_times = []
        new_freqs = []

        for offset in time_offsets:
            t = actual_merger_time - offset  # Time before merger
            if t < t_min:  # Only add points earlier than our existing data
                # Calculate frequency using PN formula
                f = (5.0/256.0) * ((G * M_chirp_kg / c**3)**(-5.0/8.0)) * (offset**(-3.0/8.0))

                if 20 < f < 100:  # Reasonable frequency range
                    new_times.append(t)
                    new_freqs.append(f)
                    print(f"     Added reference point: t={t:.4f}, f={f:.1f} Hz")

        # Combine with existing data
        if new_times:
            combined_times = new_times + list(inspiral_times)
            combined_freqs = new_freqs + list(inspiral_freqs)
            return combined_times, combined_freqs

    # Return original if no changes
    return inspiral_times, inspiral_freqs

# Then call this function before chirp mass estimation:

def enhanced_chirp_mass_regression(times_centered, f83_array, weights, snr):
    """
    Ensemble regression with multiple robust methods and SNR-adaptive weighting
    """
    print(f"     Applying robust ensemble regression (SNR={snr})")

    # Method 1: Weighted least squares (current method)
    wls_coeffs = np.polyfit(times_centered, f83_array, 1, w=weights)
    wls_slope, wls_intercept = wls_coeffs[0], wls_coeffs[1]

    print(f"     WLS: slope={wls_slope:.2e}, intercept={wls_intercept:.2e}")

    # Method 2 & 3: Robust regression methods (if sklearn available)
    try:
        from sklearn.linear_model import RANSACRegressor, HuberRegressor

        # Prepare data for sklearn
        X = times_centered.reshape(-1, 1)
        y = f83_array

        # RANSAC regression (outlier-resistant)
        # RANSAC regression with SNR-adaptive parameters
        print("     Applying SNR-adaptive RANSAC regression...")
        # Calculate adaptive threshold - stricter for higher SNR
        if snr > 500:
            threshold_factor = 0.2  # Much stricter for very high SNR
        elif snr > 250:
            threshold_factor = 0.3
        elif snr > 100:
            threshold_factor = 0.4
        else:
            threshold_factor = 0.5  # Original value for low SNR

        # Also adjust min_samples based on SNR
        if snr > 250:
            # For high SNR, we can be more demanding about consensus
            min_samples = max(3, len(times_centered)//2)  # Require majority consensus
        else:
            min_samples = max(2, len(times_centered)//3)  # Original setting

        ransac = RANSACRegressor(
            random_state=42,
            residual_threshold=np.std(f83_array) * threshold_factor,
            max_trials=200,  # More trials for better chance of finding optimal fit
            min_samples=min_samples,
            stop_probability=0.99  # Higher probability of finding optimal solution
        )

        ransac.fit(X, y, sample_weight=weights)
        ransac_slope = ransac.estimator_.coef_[0]
        ransac_intercept = ransac.estimator_.intercept_
        n_inliers = np.sum(ransac.inlier_mask_)

        print(f"     RANSAC: slope={ransac_slope:.2e}, intercept={ransac_intercept:.2e}")
        print(f"     RANSAC inliers: {n_inliers}/{len(times_centered)} points")

        # Huber regression (robust to noise)
        print("     Applying Huber regression...")
        huber = HuberRegressor(
            epsilon=1.35,  # Standard robust parameter
            max_iter=100,
            alpha=0.0001,  # Small regularization
            fit_intercept=True
        )

        huber.fit(X, y, sample_weight=weights)
        huber_slope = huber.coef_[0]
        huber_intercept = huber.intercept_

        print(f"     Huber: slope={huber_slope:.2e}, intercept={huber_intercept:.2e}")

        # Method quality assessment
        methods_available = 3

        # Calculate residuals for each method to assess quality
        wls_residuals = y - (wls_slope * times_centered + wls_intercept)
        ransac_residuals = y - (ransac_slope * times_centered + ransac_intercept)
        huber_residuals = y - (huber_slope * times_centered + huber_intercept)

        wls_rmse = np.sqrt(np.mean(wls_residuals**2))
        ransac_rmse = np.sqrt(np.mean(ransac_residuals**2))
        huber_rmse = np.sqrt(np.mean(huber_residuals**2))

        print(f"     Method RMSEs: WLS={wls_rmse:.2e}, RANSAC={ransac_rmse:.2e}, Huber={huber_rmse:.2e}")

        # SNR-adaptive weighting of ensemble methods
        # REPLACE WITH:
        if snr > 500:
            # Very high SNR: strongly favor robust methods with outlier detection
            print("     Very high SNR: Robust methods favored")
            wls_weight = 0.2   # Less trust in simple WLS
            ransac_weight = 0.5  # Higher weight to RANSAC for outlier handling
            huber_weight = 0.3  # Moderate weight to Huber
        elif snr > 100:
            # High SNR: balanced with preference for robustness
            print("     High SNR: Robustness-balanced ensemble")
            wls_weight = 0.3
            ransac_weight = 0.4
            huber_weight = 0.3
        elif snr > 50:
            # Medium SNR: balanced approach
            print("     Medium SNR: Balanced ensemble")
            wls_weight = 0.4
            ransac_weight = 0.3
            huber_weight = 0.3
        else:
            # Low SNR: trust robust methods more
            print("     Low SNR: Robustness-weighted ensemble")
            wls_weight = 0.2
            ransac_weight = 0.4
            huber_weight = 0.4

        # Additional quality-based weighting
        total_rmse = wls_rmse + ransac_rmse + huber_rmse
        if total_rmse > 0:
            # Give more weight to methods with lower RMSE
            rmse_weights = np.array([1/wls_rmse, 1/ransac_rmse, 1/huber_rmse])
            rmse_weights = rmse_weights / np.sum(rmse_weights)

            # Blend SNR-based and quality-based weights
            blend_factor = 0.7  # 70% SNR-based, 30% quality-based
            final_weights = (blend_factor * np.array([wls_weight, ransac_weight, huber_weight]) +
                           (1-blend_factor) * rmse_weights)

            wls_weight, ransac_weight, huber_weight = final_weights

        # Calculate ensemble result
        final_slope = (wls_weight * wls_slope +
                      ransac_weight * ransac_slope +
                      huber_weight * huber_slope)

        final_intercept = (wls_weight * wls_intercept +
                          ransac_weight * ransac_intercept +
                          huber_weight * huber_intercept)

        print(f"     Final weights: WLS={wls_weight:.2f}, RANSAC={ransac_weight:.2f}, Huber={huber_weight:.2f}")
        print(f"     Ensemble result: slope={final_slope:.2e}, intercept={final_intercept:.2e}")

        # Quality metrics for the ensemble
        ensemble_residuals = y - (final_slope * times_centered + final_intercept)
        ensemble_rmse = np.sqrt(np.mean(ensemble_residuals**2))
        ensemble_r2 = 1 - np.sum(ensemble_residuals**2) / np.sum((y - np.mean(y))**2)

        print(f"     Ensemble quality: RMSE={ensemble_rmse:.2e}, R²={ensemble_r2:.3f}")

        return final_slope, final_intercept, {
            'methods_used': ['WLS', 'RANSAC', 'Huber'],
            'weights': [wls_weight, ransac_weight, huber_weight],
            'individual_rmse': [wls_rmse, ransac_rmse, huber_rmse],
            'ensemble_rmse': ensemble_rmse,
            'ensemble_r2': ensemble_r2,
            'ransac_inliers': n_inliers
        }

    except ImportError:
        print("     sklearn not available, using weighted least squares only")
        return wls_slope, wls_intercept, {
            'methods_used': ['WLS'],
            'weights': [1.0],
            'individual_rmse': [np.sqrt(np.mean((y - (wls_slope * times_centered + wls_intercept))**2))],
            'ensemble_rmse': None,
            'ensemble_r2': None,
            'ransac_inliers': None
        }

    except Exception as e:
        print(f"     Robust regression failed: {str(e)[:100]}..., using weighted least squares")
        return wls_slope, wls_intercept, {
            'methods_used': ['WLS'],
            'weights': [1.0],
            'individual_rmse': [np.sqrt(np.mean((y - (wls_slope * times_centered + wls_intercept))**2))],
            'ensemble_rmse': None,
            'ensemble_r2': None,
            'ransac_inliers': None
        }


def calculate_enhanced_weights(clean_times, clean_freqs, snr, actual_merger_time):
    """
    Enhanced weighting scheme with multiple factors for optimal chirp mass estimation
    """
    n_points = len(clean_times)
    print(f"     Calculating enhanced weights for {n_points} data points")

    # 1. Time-based weights (proximity to merger)
    print("     Computing time-based weights...")
    time_to_merger = actual_merger_time - clean_times

    # Exponential preference for later times (closer to merger)
    # More aggressive weighting for better signal quality near merger
    time_decay = 0.1 if snr > 50 else 0.15  # Adaptive decay based on SNR
    time_weights = np.exp(-time_to_merger / time_decay)

    # Normalize time weights
    time_weights = time_weights / np.max(time_weights)

    # 2. Frequency-based weights (LIGO sensitivity band optimization)
    print("     Computing frequency-based weights...")
    freq_weights = np.ones_like(clean_freqs)

    # LIGO's most sensitive frequency band (100-300 Hz)
    optimal_mask = (clean_freqs >= 100) & (clean_freqs <= 300)
    freq_weights[optimal_mask] *= 1.5

    # Good sensitivity band (50-100 Hz and 300-400 Hz)
    good_mask = ((clean_freqs >= 50) & (clean_freqs < 100)) | \
                ((clean_freqs > 300) & (clean_freqs <= 400))
    freq_weights[good_mask] *= 1.2

    # Lower weights for extreme frequencies (poor sensitivity)
    extreme_low_mask = clean_freqs < 30
    extreme_high_mask = clean_freqs > 500
    freq_weights[extreme_low_mask] *= 0.5
    freq_weights[extreme_high_mask] *= 0.6

    # 3. SNR-dependent overall scaling
    print("     Computing SNR-dependent scaling...")
    snr_factor = min(1.0, np.log10(max(10, snr)) / np.log10(1000))

    # More aggressive SNR scaling for better dynamic range
    if snr < 30:
        base_weight = 0.2  # Conservative weighting for low SNR
    elif snr < 100:
        base_weight = 0.3 + 0.5 * snr_factor  # Gradual increase
    else:
        base_weight = 0.8 + 0.2 * snr_factor  # High confidence for high SNR

    print(f"     SNR factor: {snr_factor:.3f}, base weight: {base_weight:.3f}")

    # 4. Data quality weights (based on local signal consistency)
    print("     Computing data quality weights...")
    quality_weights = np.ones_like(clean_freqs)

    if len(clean_freqs) > 5:
        # Calculate local frequency derivatives (chirp rate)
        freq_derivs = np.gradient(clean_freqs, clean_times)

        # Robust statistics for outlier detection
        median_deriv = np.median(freq_derivs)
        mad_deriv = np.median(np.abs(freq_derivs - median_deriv))

        # Down-weight points with abnormal frequency evolution
        if mad_deriv > 0:
            # Adjust thresholds based on SNR
            if snr > 250:
                # Very strict for high SNR
                strong_threshold = 2.5
                moderate_threshold = 1.8
                strong_factor = 0.1     # More aggressive down-weighting
                moderate_factor = 0.4
            elif snr > 100:
                strong_threshold = 2.8
                moderate_threshold = 2.0
                strong_factor = 0.2
                moderate_factor = 0.5
            else:
                # Original values for low SNR
                strong_threshold = 3.0
                moderate_threshold = 2.0
                strong_factor = 0.3
                moderate_factor = 0.6

            for i, deriv in enumerate(freq_derivs):
                deviation = abs(deriv - median_deriv) / mad_deriv

                if deviation > strong_threshold:  # Strong outlier
                    # For very high deviations at high SNR, consider complete rejection
                    if snr > 200 and deviation > 10:
                        quality_weights[i] *= 0.01  # Effectively remove from consideration
                        print(f"     Extreme outlier at t={clean_times[i]:.4f}s (deviation={deviation:.1f})")
                    else:
                        quality_weights[i] *= strong_factor
                        print(f"     Strong outlier at t={clean_times[i]:.4f}s (deviation={deviation:.1f})")
                elif deviation > moderate_threshold:  # Moderate outlier
                    quality_weights[i] *= moderate_factor
                    print(f"     Moderate outlier at t={clean_times[i]:.4f}s (deviation={deviation:.1f})")

        # 5. Frequency evolution consistency check
        # Points should follow monotonic increase
        for i in range(1, len(clean_freqs)):
            if clean_freqs[i] < clean_freqs[i-1] * 0.95:  # Significant decrease
                quality_weights[i] *= 0.4
                print(f"     Non-monotonic frequency at t={clean_times[i]:.4f}s")

    # 6. Signal-to-noise local estimation
    print("     Computing local signal quality weights...")
    local_snr_weights = np.ones_like(clean_freqs)

    if len(clean_freqs) > 3:
        # Estimate local signal quality based on frequency stability
        window_size = min(3, len(clean_freqs)//3)

        for i in range(len(clean_freqs)):
            start_idx = max(0, i - window_size//2)
            end_idx = min(len(clean_freqs), i + window_size//2 + 1)
            local_freqs = clean_freqs[start_idx:end_idx]

            if len(local_freqs) > 1:
                local_stability = 1.0 / (1.0 + np.std(local_freqs) / np.mean(local_freqs))
                local_snr_weights[i] = local_stability

    # 7. Combine all weight components
    print("     Combining all weight factors...")
    combined_weights = (base_weight * time_weights * freq_weights *
                       quality_weights * local_snr_weights)

    # Ensure minimum weight (avoid zero weights)
    min_weight = 0.1
    combined_weights = np.maximum(combined_weights, min_weight)

    # Normalize weights to have mean = n_points (maintains regression scaling)
    combined_weights = combined_weights / np.mean(combined_weights) * n_points

    # Quality diagnostics
    weight_ratio = np.max(combined_weights) / np.min(combined_weights)
    effective_points = np.sum(combined_weights)**2 / np.sum(combined_weights**2)

    print(f"     Weight statistics:")
    print(f"       Range: {np.min(combined_weights):.2f} - {np.max(combined_weights):.2f}")
    print(f"       Ratio: {weight_ratio:.1f}:1")
    print(f"       Effective points: {effective_points:.1f}/{n_points}")

    return combined_weights

def enhanced_quality_assessment(chirp_result, times_centered, f83_array, snr, clean_times, clean_freqs):
    """
    Comprehensive quality assessment with multiple validation layers
    """
    print("     Performing comprehensive quality assessment...")
    quality_metrics = {}

    # 1. Statistical Quality Assessment
    r_squared = chirp_result.get('r_squared', 0)
    quality_metrics['r_squared'] = r_squared

    # R-squared scoring
    if r_squared > 0.9:
        r2_score = 1.0
    elif r_squared > 0.8:
        r2_score = 0.9
    elif r_squared > 0.7:
        r2_score = 0.7
    elif r_squared > 0.5:
        r2_score = 0.5
    else:
        r2_score = 0.2

    quality_metrics['r2_score'] = r2_score
    print(f"       R² quality: {r_squared:.3f} → score: {r2_score:.2f}")

    # 2. Physical Plausibility Assessment
    estimated_mass = chirp_result.get('chirp_mass', 0)

    if 5 < estimated_mass < 150:
        mass_plausibility = 1.0  # Typical stellar-mass range
        plausibility_desc = "excellent"
    elif 3 < estimated_mass < 200:
        mass_plausibility = 0.7  # Extended but reasonable range
        plausibility_desc = "good"
    elif 1 < estimated_mass < 300:
        mass_plausibility = 0.4  # Marginal range
        plausibility_desc = "marginal"
    else:
        mass_plausibility = 0.1  # Unphysical range
        plausibility_desc = "poor"

    quality_metrics['mass_plausibility'] = mass_plausibility
    print(f"       Mass plausibility: {estimated_mass:.1f}M☉ → {plausibility_desc} ({mass_plausibility:.2f})")

    # 3. Data Quantity and Coverage Assessment
    n_points = len(times_centered)
    time_span = max(times_centered) - min(times_centered) if len(times_centered) > 1 else 0

    # Data quantity score
    if n_points >= 10:
        quantity_score = 1.0
    elif n_points >= 7:
        quantity_score = 0.8
    elif n_points >= 5:
        quantity_score = 0.6
    elif n_points >= 3:
        quantity_score = 0.4
    else:
        quantity_score = 0.2

    quality_metrics['data_quantity'] = quantity_score

    # Time coverage score (optimal at 100ms+ span)
    optimal_span = 0.1  # 100ms
    coverage_score = min(1.0, time_span / optimal_span)
    quality_metrics['time_coverage'] = coverage_score

    print(f"       Data quality: {n_points} points, {time_span*1000:.1f}ms span")
    print(f"       Quantity score: {quantity_score:.2f}, Coverage score: {coverage_score:.2f}")

    # 4. Residual Analysis for Systematic Trends
    if len(times_centered) > 2 and chirp_result.get('slope') and chirp_result.get('intercept'):
        slope = chirp_result['slope']
        intercept = chirp_result['intercept']
        predicted = slope * times_centered + intercept
        residuals = f83_array - predicted

        # Check for systematic trends in residuals
        if len(times_centered) > 2:
            try:
                residual_trend = np.corrcoef(times_centered, residuals)[0,1]
                # High correlation indicates systematic bias
                trend_score = max(0, 1 - 5*abs(residual_trend))
            except:
                trend_score = 0.5
        else:
            trend_score = 0.5

        # Residual normality check (simplified)
        residual_std = np.std(residuals)
        predicted_std = np.std(predicted)
        if predicted_std > 0:
            noise_ratio = residual_std / predicted_std
            noise_score = max(0, min(1, 2 - noise_ratio))  # Good if ratio < 1
        else:
            noise_score = 0.5

        quality_metrics['residual_randomness'] = trend_score
        quality_metrics['noise_level'] = noise_score

        print(f"       Residual analysis: trend score={trend_score:.2f}, noise score={noise_score:.2f}")
    else:
        quality_metrics['residual_randomness'] = 0.5
        quality_metrics['noise_level'] = 0.5
        print(f"       Residual analysis: insufficient data")

    # 5. SNR Consistency Check
    # Compare expected error with actual uncertainty
    expected_error_percent = max(5, 50 / np.sqrt(max(1, snr/20)))  # Empirical scaling
    actual_error_percent = (chirp_result.get('chirp_mass_error', estimated_mass) /
                           max(estimated_mass, 1)) * 100

    if actual_error_percent > 0:
        snr_consistency = min(1.0, expected_error_percent / actual_error_percent)
    else:
        snr_consistency = 0.5

    quality_metrics['snr_consistency'] = snr_consistency
    print(f"       SNR consistency: expected {expected_error_percent:.1f}%, actual {actual_error_percent:.1f}% → score: {snr_consistency:.2f}")

    # 6. Frequency Evolution Quality
    if len(clean_freqs) > 3:
        # Check monotonicity
        freq_increases = np.diff(clean_freqs) > 0
        monotonicity = np.mean(freq_increases)

        # Check chirp rate consistency
        if len(clean_freqs) > 2:
            chirp_rates = np.diff(clean_freqs) / np.diff(clean_times)
            rate_consistency = 1.0 / (1.0 + np.std(chirp_rates) / max(np.mean(chirp_rates), 1))
        else:
            rate_consistency = 0.5

        freq_quality = 0.6 * monotonicity + 0.4 * rate_consistency
        quality_metrics['frequency_evolution'] = freq_quality

        print(f"       Frequency evolution: monotonicity={monotonicity:.2f}, consistency={rate_consistency:.2f} → score: {freq_quality:.2f}")
    else:
        quality_metrics['frequency_evolution'] = 0.5
        print(f"       Frequency evolution: insufficient data")

    # 7. Composite Quality Score
    # Weighted combination of all quality metrics
    weights = {
        'r2_score': 0.25,           # Statistical fit quality
        'mass_plausibility': 0.20,  # Physical reasonableness
        'data_quantity': 0.15,      # Sufficient data points
        'time_coverage': 0.10,      # Adequate time span
        'residual_randomness': 0.10, # No systematic bias
        'noise_level': 0.05,        # Good signal-to-noise
        'snr_consistency': 0.10,    # Consistent with SNR expectations
        'frequency_evolution': 0.05  # Good frequency data
    }

    composite_score = sum(weights[metric] * score for metric, score in quality_metrics.items()
                         if metric in weights)

    # Ensure score is between 0 and 1
    composite_score = max(0.0, min(1.0, composite_score))
    quality_metrics['composite_quality'] = composite_score

    # Quality classification
    if composite_score > 0.8:
        quality_class = "EXCELLENT"
    elif composite_score > 0.6:
        quality_class = "GOOD"
    elif composite_score > 0.4:
        quality_class = "FAIR"
    elif composite_score > 0.2:
        quality_class = "POOR"
    else:
        quality_class = "VERY_POOR"

    quality_metrics['quality_class'] = quality_class

    print(f"       Composite quality: {composite_score:.3f} ({quality_class})")

    return quality_metrics



def estimate_chirp_mass_simple(inspiral_times, inspiral_freqs, snr, actual_merger_time):
    """
    Improved chirp mass estimation with simplified filtering, robust validation, and comprehensive error analysis.
    Consolidates all improvements into a single function while maintaining the original interface.
    """
    import numpy as np
    from scipy import stats

    # ================================================================
    # INPUT VALIDATION AND PREPROCESSING
    # ================================================================

    if len(inspiral_times) != len(inspiral_freqs):
        print(f"WARNING: Input array length mismatch! times: {len(inspiral_times)}, freqs: {len(inspiral_freqs)}")
        min_len = min(len(inspiral_times), len(inspiral_freqs))
        inspiral_times = inspiral_times[:min_len]
        inspiral_freqs = inspiral_freqs[:min_len]

    if len(inspiral_times) < 3:
        print("ERROR: Insufficient data points for chirp mass estimation")
        return _create_empty_chirp_result()

    # Convert to numpy arrays and sort by time
    times_array = np.array(inspiral_times)
    freqs_array = np.array(inspiral_freqs)
    sort_idx = np.argsort(times_array)
    times_array = times_array[sort_idx]
    freqs_array = freqs_array[sort_idx]

    # ================================================================
    # SIMPLIFIED PHYSICS-BASED FILTERING
    # ================================================================

    # Calculate SNR factor for adaptive parameters
    snr_factor = min(1.0, np.log10(max(10, snr)) / np.log10(1000))

    # Physics-based frequency bounds (simplified from complex adaptive filtering)
    f_min, f_max = 15.0, 300 + 200 * snr_factor  # 15-300 Hz at low SNR, 15-500 Hz at high SNR
    frequency_mask = (freqs_array >= f_min) & (freqs_array <= f_max)

    # Simple monotonicity check - frequency should generally increase
    monotonic_mask = np.ones(len(freqs_array), dtype=bool)
    for i in range(1, len(freqs_array)):
        if freqs_array[i] < freqs_array[i-1] * 0.95:  # Allow 5% decrease for noise
            monotonic_mask[i] = False

    # Apply filtering
    valid_mask = frequency_mask & monotonic_mask
    clean_times = times_array[valid_mask]
    clean_freqs = freqs_array[valid_mask]

    print(f"     Filtering: {len(times_array)} → {len(clean_times)} points "
          f"(removed {np.sum(~valid_mask)} outliers)")

    if len(clean_times) < 3:
        print("ERROR: Too few points after filtering")
        return _create_empty_chirp_result()

    # ================================================================
    # ROBUST OUTLIER DETECTION WITH MAD
    # ================================================================

    # Convert to f^(-8/3) and center times for numerical stability
    f83_array = clean_freqs**(-8/3)
    t_center = clean_times[0]
    times_centered = clean_times - t_center

    # Initial fit for outlier detection
    initial_fit = np.polyfit(times_centered, f83_array, 1)
    initial_pred = initial_fit[0] * times_centered + initial_fit[1]
    residuals = f83_array - initial_pred

    # MAD-based outlier detection (more robust than standard deviation)
    mad = np.median(np.abs(residuals - np.median(residuals)))
    threshold = 4.0 * 1.4826 * mad  # ~3 sigma equivalent, slightly more permissive
    good_indices = np.abs(residuals) < threshold

    # Remove outliers if found and we still have enough points
    if np.sum(~good_indices) > 0 and np.sum(good_indices) >= 3:
        print(f"     Removing {np.sum(~good_indices)}/{len(good_indices)} outliers using MAD detection")
        times_centered = times_centered[good_indices]
        f83_array = f83_array[good_indices]
        clean_times = clean_times[good_indices]
        clean_freqs = clean_freqs[good_indices]

    n = len(times_centered)

    # ================================================================
    # WEIGHTED REGRESSION WITH IMPROVED WEIGHTING
    # ================================================================

    # Create weights - balanced between time proximity to merger and data quality
    # Replace the basic weighting calculation with enhanced weighting:
    print("     Applying enhanced multi-factor weighting strategy...")
    weights = calculate_enhanced_weights(clean_times, clean_freqs, snr, actual_merger_time)

    print(f"     Enhanced weighting complete: mean weight = {np.mean(weights):.2f}")

            # Replace the simple regression with robust ensemble:
    print("     Performing robust ensemble regression...")
    slope, intercept, regression_info = enhanced_chirp_mass_regression(
        times_centered, f83_array, weights, snr
    )

    # Store regression information for quality assessment
    regression_quality = regression_info

    # ================================================================
    # PHYSICS VALIDATION
    # ================================================================

    # For chirps, f^(-8/3) vs time must have negative slope
    if slope >= 0:
        print("ERROR: Positive slope in f^(-8/3) vs time - physically impossible for a chirp")
        return _create_empty_chirp_result()

    # Calculate R-squared
    pred_vals = slope * times_centered + intercept
    residuals = f83_array - pred_vals
    ss_res = np.sum(residuals**2)
    ss_tot = np.sum((f83_array - np.mean(f83_array))**2)
    r_squared = max(0, 1 - (ss_res / ss_tot)) if ss_tot > 0 else 0

    # ================================================================
    # ENHANCED BOOTSTRAP ERROR ESTIMATION
    # ================================================================

    # Physical constants
    G = 6.67430e-11  # m^3 kg^-1 s^-2
    c = 299792458.0  # m/s
    pi = np.pi
    M_sun = 1.989e30  # kg
    conversion_factor = (5/256) * (pi)**(-8/3)

    # Primary chirp mass
    M_chirp = ((-slope) * conversion_factor)**(3/5) * (c**3/G) / M_sun

    # Bootstrap resampling with improved convergence checking
    bootstrap_samples = 1000
    min_success_rate = 0.6  # Require 60% success rate
    bootstrap_masses = []
    bootstrap_slopes = []

    for i in range(bootstrap_samples):
        # Bootstrap resample
        indices = np.random.choice(n, size=n, replace=True)
        bs_t = times_centered[indices]
        bs_f83 = f83_array[indices]
        bs_w = weights[indices]

        try:
            bs_coeffs = np.polyfit(bs_t, bs_f83, 1, w=bs_w)
            bs_slope = bs_coeffs[0]

            # Only accept physical results
            if bs_slope < 0:
                bootstrap_slopes.append(bs_slope)
                bs_mass = ((-bs_slope) * conversion_factor)**(3/5) * (c**3/G) / M_sun
                bootstrap_masses.append(bs_mass)
        except:
            continue

    # Check bootstrap success and convergence
    success_rate = len(bootstrap_slopes) / bootstrap_samples
    print(f"     Bootstrap success rate: {success_rate:.1%}")

    # Initialize error variables
    M_chirp_err = None
    M_chirp_err_lower = None
    M_chirp_err_upper = None
    slope_error = None
    bootstrap_convergence = None

    if success_rate >= min_success_rate:
        bootstrap_masses = np.array(bootstrap_masses)
        bootstrap_slopes = np.array(bootstrap_slopes)

        # Convergence check
        if len(bootstrap_masses) > 100:
            half_point = len(bootstrap_masses) // 2
            first_half_std = np.std(bootstrap_masses[:half_point])
            second_half_std = np.std(bootstrap_masses[half_point:])
            full_std = np.std(bootstrap_masses)

            bootstrap_convergence = abs(first_half_std - second_half_std) / full_std if full_std > 0 else 0
            print(f"     Bootstrap convergence: {bootstrap_convergence:.3f} "
                  f"({'GOOD' if bootstrap_convergence < 0.1 else 'FAIR' if bootstrap_convergence < 0.2 else 'POOR'})")

        # Calculate asymmetric errors from percentiles
        mass_percentiles = np.percentile(bootstrap_masses, [16, 50, 84])
        M_chirp_err_lower = mass_percentiles[1] - mass_percentiles[0]
        M_chirp_err_upper = mass_percentiles[2] - mass_percentiles[1]
        M_chirp_err = 0.5 * (M_chirp_err_lower + M_chirp_err_upper)

        slope_error = np.std(bootstrap_slopes)

    else:
        # Fallback to jackknife method
        print(f"     WARNING: Low bootstrap success rate, using jackknife method")
        jackknife_slopes = []
        for i in range(n):
            mask = np.ones(n, dtype=bool)
            mask[i] = False
            try:
                jk_coeffs = np.polyfit(times_centered[mask], f83_array[mask], 1, w=weights[mask])
                if jk_coeffs[0] < 0:  # Only accept physical slopes
                    jackknife_slopes.append(jk_coeffs[0])
            except:
                continue

        if len(jackknife_slopes) >= 0.5 * n:  # Need at least 50% success
            jackknife_std = np.std(jackknife_slopes) * np.sqrt(n-1)
            rel_err = jackknife_std / (-slope)
            M_chirp_err = M_chirp * (3/5) * rel_err
            slope_error = jackknife_std

            # Symmetric errors for jackknife
            M_chirp_err_lower = M_chirp_err
            M_chirp_err_upper = M_chirp_err
        else:
            # Final fallback - analytical error estimate
            if n > 2:
                x_var = np.sum((times_centered - np.mean(times_centered))**2)
                analytical_std = np.sqrt(ss_res / (n - 2)) / np.sqrt(x_var) if x_var > 0 else abs(slope) * 0.3
            else:
                analytical_std = abs(slope) * 0.3

            rel_err = analytical_std / (-slope)
            M_chirp_err = M_chirp * (3/5) * rel_err
            slope_error = analytical_std

            # Symmetric errors for analytical method
            M_chirp_err_lower = M_chirp_err
            M_chirp_err_upper = M_chirp_err

    # ================================================================
    # SNR-BASED ERROR FLOORS
    # ================================================================

    # Apply minimum error based on SNR (prevent overconfidence)
    if snr < 20:
        min_error_fraction = 0.30  # 30% minimum at low SNR
    elif snr < 50:
        min_error_fraction = 0.20  # 20% minimum at medium SNR
    elif snr < 100:
        min_error_fraction = 0.10  # 10% minimum at high SNR
    else:
        min_error_fraction = 0.05  # 5% minimum at very high SNR

    min_error = M_chirp * min_error_fraction
    M_chirp_err = max(M_chirp_err or 0, min_error)
    M_chirp_err_lower = max(M_chirp_err_lower or 0, min_error)
    M_chirp_err_upper = max(M_chirp_err_upper or 0, min_error)

    # ================================================================
    # SYSTEMATIC VALIDATION (if possible)
    # ================================================================

    validation_info = {}

    # ================================================================
    # QUALITY ASSESSMENT
    # ================================================================

    quality_score = _calculate_chirp_quality_score(r_squared, success_rate, bootstrap_convergence, n, snr)

    print(f"     Final chirp mass: {M_chirp:.2f} +{M_chirp_err_upper:.2f} -{M_chirp_err_lower:.2f} M☉")
    print(f"     Quality score: {quality_score:.2f}/1.0, R² = {r_squared:.3f}")

    # ================================================================
    # RETURN COMPREHENSIVE RESULTS
    # ================================================================

    # Ensure all arrays are consistent length
    min_len = min(len(clean_times), len(f83_array), len(weights))

    # Ensure all arrays are consistent length
    # Add this right before your return statement:
    print("     Comprehensive quality assessment...")
    validation_info = enhanced_quality_assessment(
        {'chirp_mass': M_chirp, 'chirp_mass_error': M_chirp_err,
        'r_squared': r_squared, 'slope': slope, 'intercept': intercept},
        times_centered, f83_array, snr, clean_times, clean_freqs
    )

    quality_score = validation_info['composite_quality']
    quality_class = validation_info['quality_class']

    print(f"     Final quality assessment: {quality_score:.3f} ({quality_class})")


    return {
        'chirp_mass': M_chirp,
        'chirp_mass_error': M_chirp_err,
        'slope': slope,
        'intercept': intercept,
        'slope_error': slope_error,
        'r_squared': r_squared,
        'asymmetric_errors': (M_chirp_err_lower, M_chirp_err_upper),
        'bootstrap_convergence': bootstrap_convergence,
        'bootstrap_samples': len(bootstrap_masses) if 'bootstrap_masses' in locals() else 0,
        'bootstrap_success_rate': success_rate if 'success_rate' in locals() else 0.0,
        'quality_score': quality_score,
        'n_data_points': n,
        'snr_factor': snr_factor,
        'validation': validation_info,
        'times_array': clean_times[:min_len],
        'f83_array': f83_array[:min_len],
        'weights': weights[:min_len],
        't_center': t_center,  # ← ADD THIS LINE - store original t_center!
        'regression_quality': regression_quality,  # Add this line
        'quality_score': quality_score,
        'validation': validation_info,
        'quality_class': quality_class
    }


def validate_pre_merger_data_quality(segment_times, segment_values, actual_merger_time, snr):
    """
    For high SNR cases, validate that data quality is sufficient in the
    critical pre-merger period where frequency changes rapidly.
    """
    if snr < 100:
        return True  # Skip for low SNR

    # Define critical zone (last ~20ms before merger)
    critical_start = actual_merger_time - 0.020
    critical_end = actual_merger_time

    # Extract data in critical zone
    critical_indices = [i for i, t in enumerate(segment_times)
                       if critical_start <= t <= critical_end]

    if len(critical_indices) < 10:
        print("     WARNING: Insufficient data points in critical pre-merger zone")
        return False

    # For high SNR, check signal amplitude in critical region
    critical_values = [segment_values[i] for i in critical_indices]
    amplitude = np.max(np.abs(critical_values))

    # Check zero crossings density in critical region
    zero_crossings = 0
    for i in range(1, len(critical_values)):
        if critical_values[i-1] * critical_values[i] <= 0:
            zero_crossings += 1

    # Calculate expected crossings (rough estimate)
    expected_crossings = 4  # At least 4 in last 20ms

    if zero_crossings < expected_crossings:
        print(f"     WARNING: Found only {zero_crossings} zero crossings in pre-merger zone")
        print(f"     Expected at least {expected_crossings} for SNR={snr}")
        return False

    return True

def _create_empty_chirp_result():
    """Helper function to return empty result when chirp mass analysis fails"""
    return {
        'chirp_mass': None,
        'chirp_mass_error': None,
        'slope': None,
        'intercept': None,
        'slope_error': None,
        'r_squared': 0.0,
        'asymmetric_errors': (None, None),
        'bootstrap_convergence': None,
        'bootstrap_samples': 0,
        'bootstrap_success_rate': 0.0,
        'quality_score': 0.0,
        'n_data_points': 0,
        'snr_factor': 0.0,
        'validation': {},
        'times_array': np.array([]),
        'f83_array': np.array([]),
        'weights': np.array([])
    }


def _calculate_chirp_quality_score(r_squared, success_rate, convergence_metric, n_points, snr):
    """Calculate overall quality score for chirp mass estimation (0-1)"""
    # R-squared component (0-1)
    r2_score = min(1.0, max(0.0, r_squared))

    # Bootstrap success component (0-1)
    success_score = min(1.0, max(0.0, success_rate or 0))

    # Convergence component (0-1)
    if convergence_metric is not None:
        convergence_score = max(0.0, min(1.0, 1.0 - convergence_metric * 5))
    else:
        convergence_score = 0.5  # Neutral if not available

    # Data quantity component (0-1)
    data_score = min(1.0, n_points / 15)  # Normalize to 15 points

    # SNR component (0-1)
    snr_score = min(1.0, np.log10(max(10, snr)) / np.log10(1000))

    # Weighted combination
    quality = (0.3 * r2_score + 0.25 * success_score + 0.2 * convergence_score +
               0.15 * data_score + 0.1 * snr_score)

    return min(1.0, max(0.0, quality))

# -----------------------------------------------------------------------------
# SECTION 2: PIPELINE CONFIGURATION
# -----------------------------------------------------------------------------

GPS_START_TIME = 1242442951
DURATION = 32
SAMPLE_RATE = 4096

# # Add current date, time, and user information
# current_datetime = datetime.datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S")
# current_user = "1Aadi0"
# print(f"Pipeline execution: {current_datetime} UTC | User: {current_user}")

# Get user input for parameters
params = get_initial_parameters()
params['sample_rate'] = SAMPLE_RATE  # Ensure sample rate is set correctly

# -----------------------------------------------------------------------------
# SECTION 3: RUN THE CORRECTED PIPELINE
# -----------------------------------------------------------------------------

print("--- Starting Gravitational Wave Analysis Pipeline ---")

# Step 1: Fetch the real LIGO noise data
print("\n--> Step 1: Fetching real LIGO noise data...")
noise_data = timeseries.TimeSeries.fetch_open_data('H1', int(GPS_START_TIME), int(GPS_START_TIME + DURATION))
noise_data = noise_data.to_pycbc()

# Step 2: Condition the data AND Estimate the Power Spectral Density (PSD) FIRST
print("\n--> Step 2: Conditioning data and Estimating REAL Power Spectral Density (PSD)...")
strain_highpassed = highpass(noise_data, 15.0)
strain_resampled = resample_to_delta_t(strain_highpassed, 1.0 / SAMPLE_RATE)
conditioned_data_pre_injection = strain_resampled.crop(2, 2)

psd = conditioned_data_pre_injection.psd(4)
psd = interpolate(psd, conditioned_data_pre_injection.delta_f)
psd = inverse_spectrum_truncation(psd, int(4 * conditioned_data_pre_injection.sample_rate), low_frequency_cutoff=15)
print("     Real PSD estimation complete.")

# Step 3: NOW Inject the signal using the real data and real PSD
print("\n--> Step 3: Injecting signal with corrected scaling...")
injected_strain_data, template_signal, injection_time = inject_signal_corrected(
    real_noise_data=conditioned_data_pre_injection,
    real_psd=psd,
    params=params
)

# ### <<< PLOT ADDED >>>
# Step 3.1: Visualize the Raw Injected Data
print("\n--> Plotting raw data to verify injection (signal should be hidden in noise)...")
plt.figure(figsize=(15, 4))
plt.plot(injected_strain_data.sample_times, injected_strain_data, label='Raw H1 Strain')
plt.title("Raw Strain Data with Injected Signal (Signal is not visible)")
plt.xlabel("Time (s)")
plt.ylabel("Strain")
plt.xlim(injection_time - 10, injection_time + 10)
plt.axvline(injection_time, color='red', linestyle='--', label=f'Injection Time ({injection_time:.2f})')
plt.legend()
plt.grid(True)
plt.show()

# ### <<< PLOT ADDED >>>
# Step 3.2: Enhanced Multi-Resolution Signal Preprocessing
print("\n--> Enhanced signal preprocessing with wavelet denoising...")

def enhanced_signal_preprocessing(injected_strain_data, psd, snr):
    """
    Multi-scale signal enhancement with SNR-adaptive wavelet denoising
    """
    import pywt
    from scipy import signal

    print(f"     Applying SNR-adaptive preprocessing (SNR={snr})")

    # Step 1: Standard whitening first
    white_data = (injected_strain_data.to_frequencyseries() / psd**0.5).to_timeseries()

    # Step 2: SNR-adaptive wavelet denoising
    if snr < 50:
        print("     Low SNR detected: Applying aggressive wavelet denoising")
        # Aggressive denoising for low SNR
        try:
            coeffs = pywt.wavedec(white_data.numpy(), 'db8', level=6)
            # Adaptive threshold based on noise characteristics
            sigma = np.median(np.abs(coeffs[-1])) / 0.6745  # Robust noise estimate
            threshold = sigma * np.sqrt(2 * np.log(len(white_data.numpy())))

            # Apply soft thresholding
            coeffs_thresh = [pywt.threshold(c, threshold * 0.1, mode='soft') for c in coeffs]
            denoised_signal = pywt.waverec(coeffs_thresh, 'db8')

            # Reconstruct TimeSeries object
            from pycbc.types import TimeSeries
            white_data = TimeSeries(denoised_signal[:len(white_data)],
                                   delta_t=white_data.delta_t,
                                   epoch=white_data.start_time)
            print(f"     Wavelet denoising applied (threshold: {threshold:.2e})")

        except Exception as e:
            print(f"     Wavelet denoising failed: {e}, using standard whitening")

    elif snr < 100:
        print("     Medium SNR detected: Applying moderate denoising")
        # Light denoising for medium SNR
        try:
            coeffs = pywt.wavedec(white_data.numpy(), 'db4', level=4)
            sigma = np.median(np.abs(coeffs[-1])) / 0.6745
            threshold = sigma * np.sqrt(2 * np.log(len(white_data.numpy())))

            # Lighter thresholding
            coeffs_thresh = [pywt.threshold(c, threshold * 0.05, mode='soft') for c in coeffs]
            denoised_signal = pywt.waverec(coeffs_thresh, 'db4')

            from pycbc.types import TimeSeries
            white_data = TimeSeries(denoised_signal[:len(white_data)],
                                   delta_t=white_data.delta_t,
                                   epoch=white_data.start_time)
            print(f"     Light wavelet denoising applied")

        except Exception as e:
            print(f"     Light denoising failed: {e}, using standard whitening")
    else:
        print("     High SNR detected: Minimal preprocessing")

    # Step 3: SNR-adaptive bandpass filtering
    if snr < 30:
        # Narrower band for low SNR - focus on most sensitive frequencies
        print("     Applying narrow bandpass (35-250 Hz) for low SNR")
        white_data_bp = white_data.highpass_fir(35., 512).lowpass_fir(250, 512)
    elif snr < 100:
        # Standard band for medium SNR
        print("     Applying standard bandpass (30-300 Hz) for medium SNR")
        white_data_bp = white_data.highpass_fir(30., 512).lowpass_fir(300, 512)
    else:
        # Wider band for high SNR - can handle more frequency content
        print("     Applying wide bandpass (25-400 Hz) for high SNR")
        white_data_bp = white_data.highpass_fir(25., 512).lowpass_fir(400, 512)

    # Step 4: Optional: Spectral whitening enhancement for very low SNR
    if snr < 20:
        print("     Applying additional spectral whitening for very low SNR")
        try:
            # Calculate local PSD and apply additional whitening
            local_psd = white_data_bp.psd(2)
            from pycbc.psd import interpolate
            local_psd = interpolate(local_psd, white_data_bp.delta_f)

            # Apply additional whitening step
            extra_whitened = (white_data_bp.to_frequencyseries() / local_psd**0.25).to_timeseries()
            white_data_bp = extra_whitened
            print("     Additional spectral whitening applied")

        except Exception as e:
            print(f"     Additional whitening failed: {e}")

    print(f"     Enhanced preprocessing complete")
    return white_data_bp

# Apply the enhanced preprocessing
white_data_bp = enhanced_signal_preprocessing(injected_strain_data, psd, params['snr'])

print("\n--> Plotting enhanced preprocessed data...")


print("\n--> Step 3.3: Detecting actual merger time based on maximum amplitude...")
# Find the time of maximum amplitude in the whitened data
max_amp_idx = np.argmax(np.abs(white_data_bp.numpy()))
actual_merger_time = white_data_bp.sample_times[max_amp_idx]
print(f"     Theoretical injection time: {injection_time:.4f}")
print(f"     Actual merger time (max amplitude): {actual_merger_time:.4f}")
print(f"     Time difference: {(actual_merger_time - injection_time)*1000:.2f} ms")

plt.figure(figsize=(15, 4))
plt.plot(white_data_bp.sample_times, white_data_bp, label='Whitened, Bandpassed Data')
plt.title(f"Injected Signal (SNR={params['snr']}) in Whitened Data (Signal is now visible)")
plt.xlabel("Time (s)")
plt.ylabel("Whitened Strain")
plt.xlim(injection_time - 0.5, injection_time + 0.5)
plt.axvline(injection_time, color='red', linestyle='--',
            label=f'Injection Time ({injection_time:.2f})')
plt.axvline(actual_merger_time, color='green', linestyle='--',
            label=f'Actual Merger (Max Amp.) ({actual_merger_time:.2f})')
plt.legend()
plt.grid(True)
plt.show()

# Step 4: Perform Matched Filtering
print("\n--> Step 4: Performing Matched Filtering...")
template_signal.resize(len(injected_strain_data))
snr_series = matched_filter(template_signal, injected_strain_data,
                            psd=psd, low_frequency_cutoff=params['f_lower'])

peak_index = np.argmax(abs(snr_series))
peak_time = snr_series.sample_times[peak_index]
peak_snr = abs(snr_series)[peak_index]

plt.figure(figsize=(15, 5))
plt.plot(snr_series.sample_times, abs(snr_series))
plt.title("Matched Filter SNR Time Series")
plt.xlabel("Time (s)"); plt.ylabel("Signal-to-Noise Ratio (SNR)")
plt.xlim(peak_time - 1, peak_time + 1)
plt.axvline(peak_time, color='red', linestyle='--', label=f'Peak SNR: {peak_snr:.2f} at {peak_time:.2f}s')
plt.legend(loc='upper left'); plt.grid(); plt.show()
print(f"Validation complete: Target SNR={params['snr']}, Recovered Peak SNR={peak_snr:.2f}")

# Step 5: Generate the Q-Transform
print("\n--> Step 5: Generating Q-transform...")
try:
    t, f, p = injected_strain_data.whiten(4, 2).qtransform(
        0.001, logfsteps=100, qrange=(8, 8), frange=(params['f_lower'], 512)
    )
    plt.figure(figsize=[15, 4])
    plt.title(f"Q-Transform of Injected Signal (SNR={params['snr']})")
    mesh = plt.pcolormesh(t, f, p**0.5, shading='auto', cmap='viridis')
    plt.colorbar(mesh, label='Whitened Strain Amplitude (√Power)')
    plt.yscale('log')
    plt.xlabel('Time (s)')
    plt.ylabel('Frequency (Hz)')
    plt.xlim(injection_time - 0.5, injection_time + 0.5)
    plt.axvline(injection_time, color='white', linestyle='--', linewidth=0.8,
                label='Injection Time')
    plt.axvline(actual_merger_time, color='cyan', linestyle='--', linewidth=0.8,
                label='Actual Merger Time')
    plt.legend(loc='upper right')
    plt.show()

except Exception as e:
    print(f"Error during Q-transform generation: {e}")

# Get user input for analysis window
window_before, window_after = find_optimal_analysis_window_snr_only(white_data_bp, actual_merger_time, params['snr'])

# Then in SECTION 7, use these values as before:
print(f"  Using dynamically optimized window: {window_before:.4f}s before and {window_after:.4f}s after merger")

# -----------------------------------------------------------------------------
# SECTION 4: Enhanced Zero-Crossing Based Frequency Estimation with COMPLETE Chirp Mass Analysis
# -----------------------------------------------------------------------------

print("\n--> Step 7: Estimating frequency using enhanced zero-crossing method...")

try:
    # First, check the available time range in the data
    data_start_time = float(white_data_bp.start_time)
    data_end_time = data_start_time + white_data_bp.duration
    print(f"     Available data time range: {data_start_time:.4f} to {data_end_time:.4f}")
    print(f"     Injection time: {injection_time:.4f}")

    # Define desired analysis window
    print(f"     Using analysis window: {window_before:.4f}s before and {window_after:.4f}s after merger")

    # Find indices corresponding to our desired range
    time_array = white_data_bp.sample_times.numpy()
    start_idx = max(0, np.searchsorted(time_array, actual_merger_time - window_before))
    end_idx = min(len(time_array) - 1, np.searchsorted(time_array, actual_merger_time + window_after))

    # Ensure we have enough data points for analysis
    if end_idx - start_idx < 100:  # Arbitrary minimum number of points
        print("     WARNING: Very few data points available for analysis")

    # Extract the relevant portion of data
    segment_times = time_array[start_idx:end_idx]
    segment_values = white_data_bp.numpy()[start_idx:end_idx]

    print(f"     Analyzing data from {segment_times[0]:.4f} to {segment_times[-1]:.4f}")
    print(f"     Number of data points: {len(segment_times)}")

    if params['snr'] > 100:
        pre_merger_quality = validate_pre_merger_data_quality(
            segment_times, segment_values, actual_merger_time, params['snr'])
        if not pre_merger_quality:
            print("     WARNING: Pre-merger data quality issues detected")
            print("     Consider manual inspection or adjusting analysis window")


    # Use the new SNR-only function to analyze zero crossings
    result = analyze_zero_crossings_snr_only(segment_times, segment_values, actual_merger_time, params['snr'])

    if result:
        zero_crossings = result['zero_crossings']
        crossing_types = result['crossing_types']
        frequencies = result['frequencies']
        crossing_times = result['crossing_times']
        crossing_phase = result['crossing_phase']
        merger_freq = result['merger_freq']

        # Use only inspiral frequencies for chirp mass estimation
        inspiral_freqs = result['inspiral_freqs']
        inspiral_times = result['inspiral_times']

        print(f"     Frequency at actual merger: {merger_freq:.1f} Hz")
        print(f"     Found {len(zero_crossings)} zero crossings ({crossing_types.count(1)} rising, {crossing_types.count(-1)} falling)")
        print(f"     Calculated {len(frequencies)} frequency points ({len(inspiral_freqs)} inspiral, {len(frequencies)-len(inspiral_freqs)} ringdown)")


        # 1. FREQUENCY CLEANING - Filter out implausible frequency jumps - ONLY on inspiral data
        # Use SNR-adaptive max frequency instead of mass-based limits
        snr_factor = min(1.0, np.log10(max(10, params['snr'])) / np.log10(1000))
        max_expected_freq = 200 + 300 * snr_factor  # 200 Hz at low SNR, 500 Hz at high SNR

        # SNR-adaptive jump ratio - more strict at high SNR
        max_jump_ratio = 4.0 - 1.0 * snr_factor  # 4.0 at low SNR, 3.0 at high SNR

        clean_inspiral_freqs = []
        clean_inspiral_times = []
        clean_inspiral_f83 = []
        # # Add this new code:
        # # Enhanced outlier detection based on local trends
        # if len(inspiral_freqs) > 3:  # Need at least 4 points to analyze trends
        #     # Calculate local slopes between consecutive points
        #     time_diffs = np.diff(inspiral_times)
        #     freq_diffs = np.diff(inspiral_freqs)
        #     freq_slopes = freq_diffs / time_diffs  # Local frequency slopes

        #     # Identify points with abnormal local slopes
        #     valid_indices = [0]  # Always include first point
        #     for i in range(len(freq_slopes)):
        #         if i > 0:
        #             # Compare with average of nearby slopes
        #             nearby_slopes = freq_slopes[max(0,i-2):i]  # Look at up to 2 previous slopes
        #             median_slope = np.median(abs(nearby_slopes)) if nearby_slopes else abs(freq_slopes[i])

        #             # More permissive at lower frequencies (earlier in inspiral)
        #             threshold = 3.0 if inspiral_freqs[i] < 50 else 2.0

        #             if abs(freq_slopes[i]) > threshold * median_slope and median_slope > 0:
        #                 print(f"     Removing point with abnormal local slope at t={inspiral_times[i+1]:.4f}")
        #             else:
        #                 valid_indices.append(i+1)  # Add this point and the next one

        #     # Use only valid indices
        #     inspiral_freqs = [inspiral_freqs[i] for i in valid_indices]
        #     inspiral_times = [inspiral_times[i] for i in valid_indices]

        # Add first frequency point if reasonable
        if inspiral_freqs and inspiral_freqs[0] < max_expected_freq:
            clean_inspiral_freqs.append(inspiral_freqs[0])
            clean_inspiral_times.append(inspiral_times[0])
            clean_inspiral_f83.append(inspiral_freqs[0]**(-8/3))

        # Apply physics-based adaptive filtering
        print("     Applying time-to-merger adaptive frequency filtering...")
        for i in range(1, len(inspiral_freqs)):
            # Skip if frequency is implausibly high
            if inspiral_freqs[i] > max_expected_freq:
                continue

            # Skip if no prior points accepted yet
            if not clean_inspiral_freqs:
                clean_inspiral_freqs.append(inspiral_freqs[i])
                clean_inspiral_times.append(inspiral_times[i])
                clean_inspiral_f83.append(inspiral_freqs[i]**(-8/3))
                continue

            # Time-to-merger based jump ratio scaling
            time_to_merger = actual_merger_time - inspiral_times[i]

            # Base jump ratio - still SNR-adaptive but less strict
            base_jump_ratio = 4.0 - 0.5 * snr_factor  # 4.0 at low SNR, 3.5 at high SNR

            # Scale ratio based on proximity to merger
            # Allow much larger jumps near merger where frequency evolves rapidly
            distance_factor = max(1.0, min(3.0, 0.1/max(0.001, time_to_merger)))
            adaptive_jump_ratio = base_jump_ratio * distance_factor

            # Check jump against adaptive threshold
            jump_ratio = max(inspiral_freqs[i]/clean_inspiral_freqs[-1],
                            clean_inspiral_freqs[-1]/inspiral_freqs[i])

            if jump_ratio > adaptive_jump_ratio:
                print(f"     Filtering jump at t={inspiral_times[i]:.4f}, ratio={jump_ratio:.1f} > {adaptive_jump_ratio:.1f}")
                continue

            # Accept this frequency point
            clean_inspiral_freqs.append(inspiral_freqs[i])
            clean_inspiral_times.append(inspiral_times[i])
            clean_inspiral_f83.append(inspiral_freqs[i]**(-8/3))

        # Report on cleaning
        original_inspiral_count = len(inspiral_freqs)
        inspiral_freqs = clean_inspiral_freqs
        inspiral_times = clean_inspiral_times
        inspiral_f_to_neg_8_3 = np.array(clean_inspiral_f83)
        inspiral_times_array = np.array(clean_inspiral_times)

        print(f"     Cleaned inspiral points: {len(inspiral_freqs)} (removed {original_inspiral_count - len(clean_inspiral_freqs)} outliers)")

        # Add theoretical reference points for high SNR cases
        if params['snr'] > 250:
            inspiral_times, inspiral_freqs = use_theoretical_reference_points(
                inspiral_times, inspiral_freqs, params, params['snr'])

            # Update derived arrays after adding theoretical points
            inspiral_f_to_neg_8_3 = np.array([f**(-8/3) for f in inspiral_freqs])
            inspiral_times_array = np.array(inspiral_times)
            print(f"     Updated with theoretical points: {len(inspiral_freqs)} total points")

        if len(inspiral_freqs) < 3:
            print("     WARNING: Not enough clean inspiral points for reliable chirp mass estimation")
            print("     Using all valid frequency points instead")
            # Fall back to using all frequency points
            all_valid_indices = []
            for i, f in enumerate(frequencies):
                if f < max_expected_freq:
                    all_valid_indices.append(i)

            if len(all_valid_indices) >= 3:
                inspiral_freqs = [frequencies[i] for i in all_valid_indices]
                inspiral_times = [crossing_times[i] for i in all_valid_indices]
                inspiral_f_to_neg_8_3 = np.array([f**(-8/3) for f in inspiral_freqs])
                inspiral_times_array = np.array(inspiral_times)
                # -------- PHYSICS-BASED CUTOFF: Only use points >15ms before merger --------
        # time_cutoff = 0.005  # 15 ms
        # mask = (actual_merger_time - np.array(inspiral_times)) > time_cutoff
        # inspiral_times = np.array(inspiral_times)[mask]
        # inspiral_freqs = np.array(inspiral_freqs)[mask]
        # # ---------------------------------------------------------------------------

        # 2. WEIGHTED REGRESSION - Using polyfit for compatibility
        if len(inspiral_freqs) >= 3:
            time_range = max(inspiral_times_array) - min(inspiral_times_array)
            if time_range > 0:
                # Less aggressive weighting scheme - more balanced between early and late points
                normalized_times = (inspiral_times_array - min(inspiral_times_array)) / time_range
                weights = np.sqrt(normalized_times) + 0.5  # Square root gives less extreme weighting

                # Ensure weights are properly normalized
                weights = weights / np.mean(weights) * len(weights)

                # Print the weight range to verify it's more balanced
                print(f"     Weight range: {min(weights):.2f} to {max(weights):.2f}")

                # Use numpy.polyfit for weighted regression
                poly_coeffs = np.polyfit(inspiral_times_array - inspiral_times_array[0],
                                        inspiral_f_to_neg_8_3,
                                        deg=1,
                                        w=weights)  # w parameter works in polyfit

                slope = poly_coeffs[0]
                intercept = poly_coeffs[1]

                # Calculate R-squared manually with better handling of edge cases
                pred_vals = slope * (inspiral_times_array - inspiral_times_array[0]) + intercept
                residuals = inspiral_f_to_neg_8_3 - pred_vals

                ss_res = np.sum(residuals**2)
                ss_tot = np.sum((inspiral_f_to_neg_8_3 - np.mean(inspiral_f_to_neg_8_3))**2)

                # Robust R-squared calculation that avoids NaN
                if ss_tot > 0:
                    r_squared = max(0, 1 - (ss_res / ss_tot))  # Ensure non-negative
                    r_value = np.sqrt(r_squared)
                else:
                    r_squared = 0
                    r_value = 0
                    print("     WARNING: Could not calculate R-squared (zero variance in data)")


                # Standard error calculation (simplified)
                n = len(inspiral_times_array)
                if n > 2:
                    x_diff = inspiral_times_array - inspiral_times_array[0]
                    std_err = np.sqrt(ss_res / (n - 2)) / np.sqrt(np.sum((x_diff - np.mean(x_diff))**2))
                else:
                    std_err = 0
            else:
                # Standard regression if time range is too small
                poly_coeffs = np.polyfit(inspiral_times_array - inspiral_times_array[0], inspiral_f_to_neg_8_3, 1)
                slope = poly_coeffs[0]
                intercept = poly_coeffs[1]
                weights = np.ones(len(inspiral_times_array))

                # Calculate R-squared value
                pred_vals = slope * (inspiral_times_array - inspiral_times_array[0]) + intercept
                residuals = inspiral_f_to_neg_8_3 - pred_vals
                ss_res = np.sum(residuals**2)
                ss_tot = np.sum((inspiral_f_to_neg_8_3 - np.mean(inspiral_f_to_neg_8_3))**2)
                r_value = np.sqrt(1 - (ss_res / ss_tot)) if ss_tot != 0 else 0

                # Standard error calculation
                n = len(inspiral_times_array)
                if n > 2:
                    x_diff = inspiral_times_array - inspiral_times_array[0]
                    std_err = np.sqrt(ss_res / (n - 2)) / np.sqrt(np.sum((x_diff - np.mean(x_diff))**2))
                else:
                    std_err = 0

            # Create prediction line for plotting
            pred_line = slope * (inspiral_times_array - inspiral_times_array[0]) + intercept

            # For binary systems, the chirp mass can be estimated from the slope
            # The correct formula from Post-Newtonian theory:
            G = 6.67430e-11  # gravitational constant in m^3 kg^-1 s^-2
            c = 299792458.0  # speed of light in m/s
            pi = np.pi
            M_sun = 1.989e30  # solar mass in kg

            # The equation relating the slope to chirp mass:
            # slope = (-256/5) * π^(8/3) * (G*M_chirp/c^3)^(5/3)

            # # CORRECTED formula - solving for M_chirp:
            # if slope < 0:  # Valid only for negative slopes as expected in GW chirps
            #     # Calculate conversion factor
            #     conversion_factor = (5/256) * (pi)**(-8/3)

            #     # Calculate chirp mass
            #     M_chirp_factor = ((-slope) * conversion_factor)**(3/5) * (c**3/G)
            #     M_chirp_solar = M_chirp_factor / M_sun

            #     # Calculate chirp mass error using error propagation
            #     relative_error_slope = std_err / (-slope)
            #     relative_error_chirp_mass = (3/5) * relative_error_slope
            #     M_chirp_error = M_chirp_solar * relative_error_chirp_mass

            #     print(f"     Estimated chirp mass: {M_chirp_solar:.2f} ± {M_chirp_error:.2f} solar masses")

            #     # For equal mass binaries, individual masses would be:
            #     if params.get('mass1') and params.get('mass2'):
            #         actual_chirp_mass = (params['mass1'] * params['mass2'])**(3/5) / (params['mass1'] + params['mass2'])**(1/5)
            #         print(f"     Actual chirp mass from parameters: {actual_chirp_mass:.2f} solar masses")
            #         print(f"     Estimation error: {(M_chirp_solar - actual_chirp_mass)/actual_chirp_mass*100:.1f}%")

            #         # Check if actual chirp mass is within error bars
            #         if abs(M_chirp_solar - actual_chirp_mass) <= M_chirp_error:
            #             print("     ✓ Actual chirp mass is within estimated error bars")
            #         else:
            #             print("     ✗ Actual chirp mass is outside estimated error bars")

            # chirp_mass_result = estimate_chirp_mass_robustly(result, actual_merger_time, params['snr'])

            # Check if robust method produced poor results
            # Replace line ~1348
# chirp_mass_result = estimate_chirp_mass_robustly(result, actual_merger_time, params['snr'])
# With:
            # Always use the simplified estimation - more reliable
            print("     Using simplified chirp mass estimation method")
            chirp_mass_result = estimate_chirp_mass_simple(inspiral_times, inspiral_freqs, params['snr'], actual_merger_time)
                # PLOTTING SECTION WITH PHASE SEPARATION
            plt.figure(figsize=(16, 16))  # Taller figure
            plt.style.use('seaborn-v0_8-whitegrid')
            mpl.rcParams.update({'font.size': 12, 'axes.labelsize': 14})

            # Top plot: Show the signal with zero crossings marked
            # Top plot: Show the signal with zero crossings marked
            # In the plotting section (around line 1340-1360):
            # Replace the current zero-crossing plotting code with:

            # Top plot: Show the signal with zero crossings marked
            # Corrected version for the first subplot
            # Create empty lists for legend handles and labels
            all_handles = []
            all_labels = []

            plt.subplot(4, 1, 1)

            # First define the rising and falling indices
            rising_indices = [i for i, t in enumerate(crossing_types) if t == 1]
            falling_indices = [i for i, t in enumerate(crossing_types) if t == -1]

            # Then plot the signal
            plt.plot(segment_times, segment_values, 'b-', linewidth=1.5, label='Whitened Signal')

            # Calculate max signal amplitude for offset positioning
            max_amp = np.max(np.abs(segment_values)) * 0.1

            # Plot rising and falling zero crossings with different markers
            if rising_indices:
                rising_times = [zero_crossings[i] for i in rising_indices]
                jittered_times = [t + np.random.uniform(-0.0002, 0.0002) for t in rising_times]
                offsets = [max_amp * (0.5 + 0.5 * np.random.random()) for _ in rising_times]
                plt.scatter(jittered_times, offsets,
                        c='red', s=80, alpha=0.8, marker='o', edgecolors='black',
                        label=f'Rising Crossings ({len(rising_indices)})')

            if falling_indices:
                falling_times = [zero_crossings[i] for i in falling_indices]
                jittered_times = [t + np.random.uniform(-0.0002, 0.0002) for t in falling_times]
                offsets = [-max_amp * (0.5 + 0.5 * np.random.random()) for _ in falling_times]
                plt.scatter(jittered_times, offsets,
                        c='green', s=80, alpha=0.8, marker='s', edgecolors='black',
                        label=f'Falling Crossings ({len(falling_indices)})')

            plt.axvline(actual_merger_time, color='green', linestyle='--', linewidth=2,
                        label=f'Merger ({actual_merger_time:.4f})')
            plt.axvline(injection_time, color='purple', linestyle='--', linewidth=2,
                        label=f'Injection ({injection_time:.4f})')
            plt.axvline(actual_merger_time + 0.001, color='orange', linestyle='-.', linewidth=2, alpha=0.7,
                    label='Ringdown Start')

            # plt.title("Signal with Zero Crossings", fontweight='bold')
            plt.xlabel("Time (s) since GPS reference")
            plt.ylabel("Whitened Strain")
            plt.grid(True)
            h1, l1 = plt.gca().get_legend_handles_labels()
            all_handles.extend(h1)
            all_labels.extend(l1)
            plt.title("Signal with Zero Crossings", fontweight='bold')

            # Second plot: Show the intervals between consecutive zero crossings with phase separation
            # Corrected version for the second subplot

            plt.subplot(4, 1, 2)

            # Calculate intervals between consecutive zero crossings
            intervals = []
            interval_times = []
            for i in range(len(zero_crossings) - 1):
                interval = zero_crossings[i+1] - zero_crossings[i]
                intervals.append(interval)
                interval_times.append((zero_crossings[i] + zero_crossings[i+1]) / 2)  # Midpoint

            # Separate inspiral and ringdown intervals
            inspiral_indices = [i for i, phase in enumerate(crossing_phase) if phase == 'inspiral']
            ringdown_indices = [i for i, phase in enumerate(crossing_phase) if phase == 'ringdown']

            # Define the inspiral and ringdown times and intervals
            inspiral_int_times = [interval_times[i] for i in inspiral_indices if i < len(interval_times)]
            inspiral_intervals = [intervals[i] for i in inspiral_indices if i < len(intervals)]

            ringdown_int_times = [interval_times[i] for i in ringdown_indices if i < len(interval_times)]
            ringdown_intervals = [intervals[i] for i in ringdown_indices if i < len(intervals)]

            # Use thicker lines and larger markers
            if inspiral_int_times:  # Check if we have any inspiral intervals
                plt.plot(inspiral_int_times, inspiral_intervals, 'm-o', linewidth=2, markersize=8,
                        label='Inspiral Intervals')

            if ringdown_int_times:  # Check if we have any ringdown intervals
                plt.plot(ringdown_int_times, ringdown_intervals, 'c--o', linewidth=1.5, markersize=6,
                        alpha=0.6, label='Ringdown (excluded)')

            plt.axvline(actual_merger_time, color='green', linestyle='--', linewidth=2,
                        label=f'Merger ({actual_merger_time:.4f})')
            plt.axvline(injection_time, color='purple', linestyle='--', linewidth=2,
                        label=f'Injection ({injection_time:.4f})')
            plt.axvline(actual_merger_time + 0.001, color='orange', linestyle='-.', linewidth=2, alpha=0.7,
                    label='Ringdown Start')

            # plt.title("Half-Period Durations Between Zero Crossings", fontweight='bold')
            plt.xlabel("Time (s) since GPS reference")
            plt.ylabel("Interval Duration (s)")
            plt.grid(True)
            h2, l2 = plt.gca().get_legend_handles_labels()
            all_handles.extend(h2)
            all_labels.extend(l2)
            plt.title("Half-Period Durations Between Zero Crossings", fontweight='bold')

            # Fourth plot: Show the calculated instantaneous frequency with phase separation
            # Replace the entire third subplot block
# Corrected version for the third subplot (4,1,3)

            plt.subplot(4, 1, 3)

            # Ensure we have proper definitions for inspiral and ringdown frequencies
            inspiral_times_plot = []
            inspiral_freqs_plot = []
            ringdown_times_plot = []
            ringdown_freqs_plot = []

            # Populate the arrays based on phase
            for i in range(len(crossing_phase)):
                if i < len(crossing_times) and i < len(frequencies):
                    if crossing_phase[i] == 'inspiral':
                        inspiral_times_plot.append(crossing_times[i])
                        inspiral_freqs_plot.append(frequencies[i])
                    elif crossing_phase[i] == 'ringdown':
                        ringdown_times_plot.append(crossing_times[i])
                        ringdown_freqs_plot.append(frequencies[i])

            # Calculate merger frequency (if not already defined)
            if 'merger_freq' not in locals() or merger_freq is None:
                # Find frequency closest to merger time
                if inspiral_times_plot and inspiral_freqs_plot:
                    merger_idx = np.argmin(np.abs(np.array(inspiral_times_plot) - actual_merger_time))
                    merger_freq = inspiral_freqs_plot[merger_idx]
                else:
                    merger_freq = 0  # Default if we can't determine

            # Use thicker lines, larger markers, and better colors
            if inspiral_times_plot:
                plt.plot(inspiral_times_plot, inspiral_freqs_plot, 'r-o', linewidth=2, markersize=8,
                        label='Inspiral Frequency')

            if ringdown_times_plot:
                plt.plot(ringdown_times_plot, ringdown_freqs_plot, 'b--o', linewidth=1.5, markersize=6,
                        alpha=0.6, label='Ringdown (excluded)')

            # Better annotation for merger frequency
            plt.annotate(f"Merger: {merger_freq:.1f} Hz",
                    xy=(actual_merger_time, merger_freq),
                    xytext=(actual_merger_time+0.005, merger_freq*1.3),
                    arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=10),
                    bbox=dict(boxstyle="round,pad=0.5", fc="yellow", ec="black", alpha=0.8),
                    fontsize=12, fontweight='bold')

            plt.axvline(actual_merger_time, color='green', linestyle='--', linewidth=2,
                        label=f'Merger ({actual_merger_time:.4f})')
            plt.axvline(injection_time, color='purple', linestyle='--', linewidth=2,
                        label=f'Injection ({injection_time:.4f})')
            plt.axvline(actual_merger_time + 0.001, color='orange', linestyle='-.', linewidth=2, alpha=0.7,
                    label='Ringdown Start')

            # plt.title("Frequency Evolution from Zero-Crossing Method", fontweight='bold')
            plt.xlabel("Time (s) since GPS reference")
            plt.ylabel("Frequency (Hz)")
            plt.grid(True)
            h3, l3 = plt.gca().get_legend_handles_labels()
            all_handles.extend(h3)
            all_labels.extend(l3)
            plt.title("Frequency Evolution from Zero-Crossing Method", fontweight='bold')

            # Fourth plot: f^(-8/3) vs time with linear fit and uncertainty band
            # Fourth plot: f^(-8/3) vs time with linear fit and uncertainty band
            # Fourth plot: f^(-8/3) vs time with linear fit and uncertainty band
            # Corrected version for the fourth subplot (4,1,4)

            plt.subplot(4, 1, 4)

            # Ensure chirp_mass_result has all required keys or provide safe defaults
            required_keys = ['times_array', 'f83_array', 'weights', 'slope', 'slope_error', 'r_squared',
                            'chirp_mass', 'chirp_mass_error']

            for key in required_keys:
                if key not in chirp_mass_result or chirp_mass_result[key] is None:
                    if key in ['times_array', 'f83_array', 'weights']:
                        chirp_mass_result[key] = np.array([])
                    elif key in ['slope', 'slope_error', 'r_squared', 'chirp_mass', 'chirp_mass_error']:
                        chirp_mass_result[key] = 0.0

            # Check if we have enough points to plot
            if len(chirp_mass_result['times_array']) > 0:
                # Use larger, more visible points with better color scheme
                plt.scatter(
                    chirp_mass_result['times_array'],
                    chirp_mass_result['f83_array'],
                    c='blue',
                    s=50 + 100 * (chirp_mass_result['weights']/np.max(chirp_mass_result['weights'])
                                if np.max(chirp_mass_result['weights']) > 0 else 1),
                    alpha=0.8,
                    marker='o',
                    edgecolors='navy',
                    label='Inspiral f^(-8/3) (size = weight)'
                )

                # Create properly defined ringdown f^(-8/3) values
                ringdown_f_to_neg_8_3 = []
                if 'ringdown_times_plot' in locals() and 'ringdown_freqs_plot' in locals():
                    if ringdown_times_plot and ringdown_freqs_plot:
                        ringdown_f_to_neg_8_3 = np.array([f**(-8/3) for f in ringdown_freqs_plot])
                        plt.scatter(ringdown_times_plot, ringdown_f_to_neg_8_3,
                                c='gray', s=30, alpha=0.4, marker='x', label='Ringdown (excluded)')

                # Create smooth regression line with clean labeling
                if len(chirp_mass_result['times_array']) > 1:
                    t_min = min(chirp_mass_result['times_array'])
                    t_max = max(chirp_mass_result['times_array'])
                    t_center = chirp_mass_result['t_center']  # ← CORRECT - use stored t_center

                    t_smooth = np.linspace(t_min, t_max, 100)

                    slope = chirp_mass_result['slope']
                    slope_err = chirp_mass_result['slope_error']
                    r_squared = chirp_mass_result['r_squared']

                    # Correct intercept: it's already at t_center
                    intercept = chirp_mass_result['intercept']

                    # Correct predicted line — always use (t - t_center)
                    y_smooth = slope * (t_smooth - t_center) + intercept

                    plt.plot(t_smooth, y_smooth, 'r-', linewidth=3,
                            label=f'Weighted Fit (R² = {r_squared:.3f})')

                    if slope_err and slope_err > 0:
                        y_upper = y_smooth + 2 * slope_err * (t_smooth - t_center)
                        y_lower = y_smooth - 2 * slope_err * (t_smooth - t_center)
                        plt.fill_between(t_smooth, y_lower, y_upper,
                                    color='r', alpha=0.3, label='95% Confidence Band')

                        band_percent = abs(2 * slope_err / slope * 100)
                        plt.annotate(f"±{band_percent:.1f}% band",
                                    xy=(t_smooth[len(t_smooth)//2], (y_upper[len(t_smooth)//2] + y_lower[len(t_smooth)//2])/2),
                                    xytext=(t_smooth[len(t_smooth)//2], y_upper[len(t_smooth)//2] * 1.2),
                                    arrowprops=dict(arrowstyle="->", color="navy"),
                                    color="navy", fontsize=9, bbox=dict(boxstyle="round", fc="white", alpha=0.7))

            # Add vertical markers
            plt.axvline(actual_merger_time, color='green', linestyle='--', linewidth=2,
                        label=f'Merger ({actual_merger_time:.4f})')
            plt.axvline(actual_merger_time + 0.001, color='orange', linestyle='-.', linewidth=2, alpha=0.7,
                    label='Ringdown Start')

            # Enhanced chirp mass information box
            chirp_mass = chirp_mass_result['chirp_mass']
            chirp_mass_err = chirp_mass_result['chirp_mass_error']
            r_squared = chirp_mass_result['r_squared']

            # Create an information box with better formatting and error handling
            # Replace your current info_text generation with this enhanced version


            if chirp_mass > 0:  # Only display if we have a valid chirp mass
                # Use asymmetric errors if available
                if 'asymmetric_errors' in chirp_mass_result and all(chirp_mass_result['asymmetric_errors']):
                    lower_err, upper_err = chirp_mass_result['asymmetric_errors']
                    err_format = f"+{upper_err:.2f} −{lower_err:.2f}"
                else:
                    err_format = f"± {chirp_mass_err:.2f}"

                # Add bootstrap convergence info if available
                bootstrap_info = ""
                if 'bootstrap_convergence' in chirp_mass_result and chirp_mass_result['bootstrap_convergence'] is not None:
                    conv = chirp_mass_result['bootstrap_convergence']
                    quality = "excellent" if conv < 0.05 else "good" if conv < 0.1 else "fair" if conv < 0.2 else "poor"
                    bootstrap_info = f"\nBootstrap: {chirp_mass_result.get('bootstrap_samples', 0)} samples, {quality} convergence"

                if params.get('mass1') and params.get('mass2'):
                    actual_chirp_mass = (params['mass1'] * params['mass2'])**(3/5) / (params['mass1'] + params['mass2'])**(1/5)

                    # Calculate percentage error
                    percent_error = abs((chirp_mass - actual_chirp_mass)/actual_chirp_mass * 100)

                    # Color code the result based on accuracy
                    if percent_error < 10:
                        box_color = 'lightgreen'
                        accuracy = 'Excellent'
                    elif percent_error < 20:
                        box_color = 'khaki'
                        accuracy = 'Good'
                    else:
                        box_color = 'lightcoral'
                        accuracy = 'Fair'

                    info_text = f"Chirp Mass: {chirp_mass:.2f} {err_format} M⊙\n"
                    info_text += f"Actual: {actual_chirp_mass:.2f} M⊙\n"
                    info_text += f"Error: {percent_error:.1f}% ({accuracy})\n"
                    info_text += f"Fit Quality: R² = {r_squared:.3f}{bootstrap_info}"
                else:
                    info_text = f"Chirp Mass: {chirp_mass:.2f} {err_format} M⊙\n"
                    info_text += f"Fit Quality: R² = {r_squared:.3f}{bootstrap_info}"
                    box_color = 'lightblue'
                # Add a more visually impressive text box
                plt.annotate(info_text,
                        xy=(0.98, 0.95),
                        xycoords='axes fraction',
                        ha='right', va='top',
                        bbox=dict(boxstyle="round,pad=0.5", fc=box_color, ec="black", alpha=0.9),
                        fontsize=12)

            # plt.title("Chirp Mass Estimation from f^(-8/3) Slope", fontweight='bold')
            plt.xlabel("Time (s) since GPS reference")
            plt.ylabel("f^(-8/3) (s^(8/3))")
            plt.grid(True)
            h4, l4 = plt.gca().get_legend_handles_labels()
            all_handles.extend(h4)
            all_labels.extend(l4)
            plt.title("Chirp Mass Estimation from f^(-8/3) Slope", fontweight='bold')

            # ADD THIS after all four subplots but before plt.show():

            # Remove duplicate entries from combined legend
            unique_labels = []
            unique_handles = []
            seen_labels = set()
            for h, l in zip(all_handles, all_labels):
                if l not in seen_labels:
                    seen_labels.add(l)
                    unique_labels.append(l)
                    unique_handles.append(h)

            # Create a single common legend for all subplots at the bottom
            plt.figlegend(unique_handles, unique_labels,
                        loc='lower center', ncol=4,
                        bbox_to_anchor=(0.5, 0),
                        fancybox=True, shadow=True, framealpha=0.9)

            # Add title and adjust layout
            plt.suptitle("Gravitational Wave Chirp Analysis - Advanced Zero-Crossing Method",
                      fontsize=20, fontweight='bold', y=0.98)

            # Add footer with analysis info
            plt.figtext(0.5, 0.01,
                    f"Analysis of signal with SNR={params['snr']}, m₁={params.get('mass1', 'N/A')}M⊙, m₂={params.get('mass2', 'N/A')}M⊙ | {current_datetime} | {current_user}",
                    ha='center', fontsize=10, style='italic')

            # Adjust subplot spacing
            plt.tight_layout(rect=[0.03, 0.08, 0.97, 0.95])

            # Additional visualization: Overlay frequency on spectrogram with zoom
            try:
                # Set zoom parameters
                zoom_time_before = 0.5  # seconds before injection time
                zoom_time_after = 0.5   # seconds after injection time

                # Calculate the time range to display
                time_min = injection_time - zoom_time_before
                time_max = injection_time + zoom_time_after

                # Create the figure with the same size as original
                plt.figure(figsize=(12, 7), facecolor='black')
                plt.title("Frequency Estimation Overlaid on Q-transform (Zoomed)", fontsize=16, fontweight='bold', color='white')
                mesh = plt.pcolormesh(t, f, p**0.5, shading='auto', cmap='viridis')
                plt.colorbar(mesh, label='Whitened Strain Amplitude').ax.yaxis.label.set_color('white')
                plt.colorbar(mesh).ax.tick_params(colors='white')

                # Plot inspiral and ringdown frequencies with different colors
                if inspiral_indices:
                    inspiral_times_mask = (np.array(inspiral_times_plot) >= min(t)) & (np.array(inspiral_times_plot) <= max(t))
                    if any(inspiral_times_mask):
                        plt.plot(np.array(inspiral_times_plot)[inspiral_times_mask],
                                np.array(inspiral_freqs_plot)[inspiral_times_mask],
                                'r-o', linewidth=2, markersize=4, label='Inspiral Frequency')

                if ringdown_indices:
                    ringdown_times_mask = (np.array(ringdown_times_plot) >= min(t)) & (np.array(ringdown_times_plot) <= max(t))
                    if any(ringdown_times_mask):
                        plt.plot(np.array(ringdown_times_plot)[ringdown_times_mask],
                                np.array(ringdown_freqs_plot)[ringdown_times_mask],
                                'b--o', linewidth=1, markersize=3, alpha=0.6, label='Ringdown (excluded)')

                # Add a vertical line marking the start of ringdown
                plt.axvline(injection_time, color='white', linestyle='--', linewidth=1.5,
               label='Injection Time')
                plt.axvline(actual_merger_time, color='cyan', linestyle='--', linewidth=1.5,
                          label='Merger Time')
                plt.axvline(actual_merger_time + 0.001, color='orange', linestyle='-.', linewidth=1.5,
                          alpha=0.7, label='Ringdown Start')

                # Apply styling
                plt.xlim(injection_time - 0.5, injection_time + 0.5)  # Appropriate zoom
                plt.yscale('log')
                plt.ylim(20, 500)
                plt.xlabel('Time (s)', color='white', fontweight='bold')
                plt.ylabel('Frequency (Hz)', color='white', fontweight='bold')
                plt.tick_params(axis='x', colors='white')
                plt.tick_params(axis='y', colors='white')
                plt.gca().set_facecolor('black')

                # MOST IMPORTANT CHANGE: Move legend outside the plot area
                plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15),
                          ncol=4, frameon=True, fancybox=True,
                          facecolor='black', edgecolor='gray', framealpha=0.8,
                          labelcolor='white')

                # Add main title and footer
                plt.suptitle("Gravitational Wave Chirp Analysis - Advanced Zero-Crossing Method",
                          fontsize=20, fontweight='bold', y=0.98, color='white')

                footer_text = f"Analysis of signal with SNR={params['snr']}, m₁={params.get('mass1', 'N/A')}M⊙, m₂={params.get('mass2', 'N/A')}M⊙ | {current_datetime} | {current_user}"
                plt.figtext(0.5, 0.01, footer_text,
                          ha='center', fontsize=10, color='white',
                          bbox=dict(boxstyle='round', fc='black', ec='gray', alpha=0.7))

                # Add extra space at bottom for the legend
                plt.tight_layout(rect=[0, 0.15, 1, 0.95])

                # # Update timestamp with current date and time
                # current_datetime = "2025-06-23 10:38:45"  # UTC
                # current_user = "1Aadi0"
                # plt.figtext(0.01, 0.01, f"{current_datetime} | {current_user}",
                #             fontsize=8, alpha=0.7)
# Add at the end of plotting code before plt.show()
# Add footer with analysis info
                # p521
                plt.show()
            except Exception as e:
                print(f"     Could not overlay frequency on spectrogram: {e}")
        else:
            print("     Not enough clean frequency points for chirp mass estimation")
    else:
        print("     Not enough zero crossings found for frequency estimation")

except Exception as e:
    print(f"     An error occurred during zero-crossing frequency estimation: {e}")
    import traceback
    traceback.print_exc()

# # Update the timestamp for the final message
# current_datetime = "2025-06-23 10:38:45"  # UTC
# current_user = "1Aadi0"
# print(f"\nEnhanced frequency analysis with complete chirp mass estimation completed at {current_datetime} by {current_user}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.cm import ScalarMappable
from scipy.stats import gaussian_kde

def create_continuous_heatmaps(results):
    """
    Create continuous heatmaps from pipeline analysis results
    with true continuous color representation for each value
    """
    # Extract data from results
    accuracy_matrix = results['accuracy_matrix']
    error_matrix = results['error_matrix']
    success_matrix = results['success_matrix']
    snr_values = results['snr_values']
    mass_values = results['mass_values']

    # Calculate necessary metrics
    mean_accuracy_vs_snr = np.mean(accuracy_matrix, axis=0)
    mean_success_vs_snr = np.mean(success_matrix, axis=0)
    mean_accuracy_vs_mass = np.mean(accuracy_matrix, axis=1)
    mean_success_vs_mass = np.mean(success_matrix, axis=1)
    successful_errors = error_matrix[success_matrix > 0]
    mean_err = np.mean(successful_errors) if len(successful_errors) > 0 else 0
    median_err = np.median(successful_errors) if len(successful_errors) > 0 else 0

    # Set up visualization style
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.rcParams.update({
        'font.size': 12,
        'font.family': 'sans-serif',
        'axes.linewidth': 1.5,
        'grid.alpha': 0.2,
        'axes.edgecolor': '#2F4F4F',
        'axes.titleweight': 'bold'
    })

    # Create figure
    fig = plt.figure(figsize=(24, 18))
    gs = fig.add_gridspec(3, 3, hspace=0.4, wspace=0.25,
                         height_ratios=[1, 1, 0.7], width_ratios=[1, 1, 1])

    # =================================================================
    # Panel 1: Accuracy Score with Continuous Gradient
    # =================================================================
    ax1 = fig.add_subplot(gs[0, 0])
    # Create continuous color mapping
    acc_norm = mcolors.Normalize(vmin=0, vmax=1)
    acc_img = ax1.imshow(accuracy_matrix,
                         cmap='RdYlGn',
                         norm=acc_norm,
                         aspect='auto',
                         interpolation='bilinear')

    # Add colorbar
    cbar1 = fig.colorbar(acc_img, ax=ax1, shrink=0.8)
    cbar1.set_label('Accuracy Score', fontweight='bold')

    # Format axes
    ax1.set_xticks(np.arange(len(snr_values)))
    ax1.set_yticks(np.arange(len(mass_values)))
    ax1.set_xticklabels([f'{snr:.0f}' for snr in snr_values])
    ax1.set_yticklabels([f'{mass:.0f}' for mass in mass_values])

    # Add value annotations
    for i in range(len(mass_values)):
        for j in range(len(snr_values)):
            ax1.text(j, i, f'{accuracy_matrix[i, j]:.2f}',
                     ha="center", va="center",
                     color="black" if accuracy_matrix[i, j] < 0.7 else "white",
                     fontsize=10)

    ax1.set_title('Continuous Accuracy Mapping\n(Each Value Has Unique Color)', fontsize=14, pad=15)
    ax1.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax1.set_ylabel('Total Mass (M☉)', fontweight='bold')

    # =================================================================
    # Panel 2: Relative Error with Continuous Gradient
    # =================================================================
    ax2 = fig.add_subplot(gs[0, 1])
    # Create continuous color mapping for errors
    err_norm = mcolors.Normalize(vmin=0, vmax=np.max(error_matrix))
    err_img = ax2.imshow(error_matrix,
                         cmap='viridis_r',
                         norm=err_norm,
                         aspect='auto',
                         interpolation='bilinear')

    # Add colorbar
    cbar2 = fig.colorbar(err_img, ax=ax2, shrink=0.8)
    cbar2.set_label('Relative Error (%)', fontweight='bold')

    # Format axes
    ax2.set_xticks(np.arange(len(snr_values)))
    ax2.set_yticks(np.arange(len(mass_values)))
    ax2.set_xticklabels([f'{snr:.0f}' for snr in snr_values])
    ax2.set_yticklabels([f'{mass:.0f}' for mass in mass_values])

    # Add value annotations
    for i in range(len(mass_values)):
        for j in range(len(snr_values)):
            ax2.text(j, i, f'{error_matrix[i, j]:.1f}',
                     ha="center", va="center",
                     color="white" if error_matrix[i, j] > 30 else "black",
                     fontsize=10)

    ax2.set_title('Continuous Error Mapping\n(Each Value Has Unique Color)', fontsize=14, pad=15)
    ax2.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax2.set_ylabel('Total Mass (M☉)', fontweight='bold')

    # =================================================================
    # Panel 3: Success Rate with Continuous Gradient
    # =================================================================
    ax3 = fig.add_subplot(gs[0, 2])
    # Create continuous color mapping
    success_norm = mcolors.Normalize(vmin=0, vmax=1)
    success_img = ax3.imshow(success_matrix,
                             cmap='Blues',
                             norm=success_norm,
                             aspect='auto',
                             interpolation='bilinear')

    # Add colorbar
    cbar3 = fig.colorbar(success_img, ax=ax3, shrink=0.8)
    cbar3.set_label('Success Rate', fontweight='bold')

    # Format axes
    ax3.set_xticks(np.arange(len(snr_values)))
    ax3.set_yticks(np.arange(len(mass_values)))
    ax3.set_xticklabels([f'{snr:.0f}' for snr in snr_values])
    ax3.set_yticklabels([f'{mass:.0f}' for mass in mass_values])

    # Add value annotations
    for i in range(len(mass_values)):
        for j in range(len(snr_values)):
            ax3.text(j, i, f'{success_matrix[i, j]:.2f}',
                     ha="center", va="center",
                     color="white" if success_matrix[i, j] > 0.6 else "black",
                     fontsize=10)

    ax3.set_title('Continuous Success Mapping\n(Each Value Has Unique Color)', fontsize=14, pad=15)
    ax3.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax3.set_ylabel('Total Mass (M☉)', fontweight='bold')

    # =================================================================
    # Panel 4: Performance vs SNR (Continuous Coloring)
    # =================================================================
    ax4 = fig.add_subplot(gs[1, 0])
    # Create continuous line with gradient coloring
    for i in range(len(snr_values)-1):
        # Create color gradient between points
        x = [snr_values[i], snr_values[i+1]]
        y_acc = [mean_accuracy_vs_snr[i], mean_accuracy_vs_snr[i+1]]
        y_success = [mean_success_vs_snr[i], mean_success_vs_snr[i+1]]

        # Interpolate color
        color_acc = plt.cm.RdYlGn(mean_accuracy_vs_snr[i])
        color_success = plt.cm.Blues(mean_success_vs_snr[i])

        ax4.plot(x, y_acc, color=color_acc, linewidth=3)
        ax4.plot(x, y_success, color=color_success, linewidth=3, linestyle='--')

    # Add markers
    ax4.scatter(snr_values, mean_accuracy_vs_snr, s=100, c=mean_accuracy_vs_snr,
               cmap='RdYlGn', edgecolor='black', zorder=10, label='Accuracy')
    ax4.scatter(snr_values, mean_success_vs_snr, s=100, c=mean_success_vs_snr,
               cmap='Blues', edgecolor='black', marker='s', zorder=10, label='Success Rate')

    # Add colorbars
    sm_acc = ScalarMappable(cmap='RdYlGn', norm=plt.Normalize(0,1))
    sm_acc.set_array([])
    cbar4a = fig.colorbar(sm_acc, ax=ax4, location='right', pad=0.01)
    cbar4a.set_label('Accuracy Score')

    sm_success = ScalarMappable(cmap='Blues', norm=plt.Normalize(0,1))
    sm_success.set_array([])
    cbar4b = fig.colorbar(sm_success, ax=ax4, location='right', pad=0.15)
    cbar4b.set_label('Success Rate')

    ax4.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax4.set_ylabel('Performance Score', fontweight='bold')
    ax4.set_title('SNR Performance Scaling with Continuous Coloring', fontsize=14)
    ax4.grid(True, alpha=0.2)
    ax4.set_xlim(snr_values[0]-5, snr_values[-1]+5)
    ax4.set_ylim(-0.05, 1.1)
    ax4.legend(loc='lower right')

    # =================================================================
    # Panel 5: Performance vs Mass (Continuous Coloring)
    # =================================================================
    ax5 = fig.add_subplot(gs[1, 1])
    # Create continuous line with gradient coloring
    for i in range(len(mass_values)-1):
        # Create color gradient between points
        x = [mass_values[i], mass_values[i+1]]
        y_acc = [mean_accuracy_vs_mass[i], mean_accuracy_vs_mass[i+1]]
        y_success = [mean_success_vs_mass[i], mean_success_vs_mass[i+1]]

        # Interpolate color
        color_acc = plt.cm.RdYlGn(mean_accuracy_vs_mass[i])
        color_success = plt.cm.Blues(mean_success_vs_mass[i])

        ax5.plot(x, y_acc, color=color_acc, linewidth=3)
        ax5.plot(x, y_success, color=color_success, linewidth=3, linestyle='--')

    # Add markers
    ax5.scatter(mass_values, mean_accuracy_vs_mass, s=100, c=mean_accuracy_vs_mass,
               cmap='RdYlGn', edgecolor='black', zorder=10, label='Accuracy')
    ax5.scatter(mass_values, mean_success_vs_mass, s=100, c=mean_success_vs_mass,
               cmap='Blues', edgecolor='black', marker='s', zorder=10, label='Success Rate')

    # Highlight optimal mass range
    ax5.axvspan(50, 70, alpha=0.15, color='gold', label='Optimal Mass Range')

    ax5.set_xlabel('Total Mass (M☉)', fontweight='bold')
    ax5.set_ylabel('Performance Score', fontweight='bold')
    ax5.set_title('Mass Sensitivity with Continuous Coloring', fontsize=14)
    ax5.grid(True, alpha=0.2)
    ax5.set_xlim(mass_values[0]-5, mass_values[-1]+5)
    ax5.set_ylim(-0.05, 1.1)
    ax5.legend(loc='lower left')

    # =================================================================
    # Panel 6: Continuous Error Distribution
    # =================================================================
    ax6 = fig.add_subplot(gs[1, 2])
    if len(successful_errors) > 0:
        # Create continuous KDE plot
        sns.kdeplot(successful_errors, ax=ax6, color='darkblue',
                   linewidth=3, alpha=0.8, label='Error Density')

        # Fill under curve with gradient
        x = np.linspace(min(successful_errors), max(successful_errors), 100)
        y = gaussian_kde(successful_errors)(x)

        # Gradient fill (simplified)
        for i in range(len(x)-1):
            ax6.fill_between([x[i], x[i+1]], [y[i], y[i+1]], alpha=0.5,
                            color=plt.cm.viridis_r(i/len(x)))

        # Add statistics lines
        ax6.axvline(mean_err, color='red', linestyle='-', linewidth=2.5,
                   label=f'Mean: {mean_err:.1f}%')
        ax6.axvline(median_err, color='green', linestyle='-', linewidth=2.5,
                   label=f'Median: {median_err:.1f}%')

    ax6.set_xlabel('Relative Error (%)', fontweight='bold')
    ax6.set_ylabel('Density', fontweight='bold')
    ax6.set_title('Continuous Error Distribution', fontsize=14)
    ax6.grid(True, alpha=0.2)
    ax6.legend()

    # =================================================================
    # Summary Panel (Enhanced)
    # =================================================================
    ax7 = fig.add_subplot(gs[2, :])
    ax7.axis('off')

    # Calculate performance metrics
    overall_accuracy = np.nanmean(accuracy_matrix[success_matrix > 0])
    overall_error = np.nanmean(error_matrix[success_matrix > 0])
    overall_success = np.mean(success_matrix)

    # Identify high-performance regions
    high_perf_mask = (accuracy_matrix > 0.7) & (success_matrix > 0.8)
    optimal_snrs = []
    optimal_masses = []

    if np.any(high_perf_mask):
        high_perf_indices = np.where(high_perf_mask)
        for i, j in zip(*high_perf_indices):
            optimal_snrs.append(snr_values[j])
            optimal_masses.append(mass_values[i])

    # Calculate optimal parameter ranges
    opt_snr_range = (min(optimal_snrs), max(optimal_snrs)) if optimal_snrs else (0, 0)
    opt_mass_range = (min(optimal_masses), max(optimal_masses)) if optimal_masses else (0, 0)

    # Create data-driven summary
    summary_text = f"""
📊 CONTINUOUS PERFORMANCE MAPPING SUMMARY

✅ Overall Success Rate: {overall_success:.1%}
🎯 Average Accuracy: {overall_accuracy:.3f}/1.000
📏 Typical Error: {overall_error:.1f}%

🔍 Key Continuous Patterns:
   • Accuracy Range: {np.min(accuracy_matrix):.2f} - {np.max(accuracy_matrix):.2f}
   • Error Range: {np.min(error_matrix):.1f}% - {np.max(error_matrix):.1f}%
   • SNR-Performance Correlation: {np.corrcoef(snr_values, mean_accuracy_vs_snr)[0,1]:.2f}

💡 Recommended Operating Parameters:
   • SNR Range: {opt_snr_range[0]:.0f}-{opt_snr_range[1]:.0f}
   • Mass Range: {opt_mass_range[0]:.0f}-{opt_mass_range[1]:.0f} M☉
   • Expected Accuracy: {np.nanmean(accuracy_matrix[high_perf_mask]):.1%} ± {np.nanstd(accuracy_matrix[high_perf_mask]):.1%}
"""

    # Create styled text box
    ax7.text(0.05, 0.5, summary_text, fontsize=14,
             bbox=dict(facecolor='#F0F8FF', alpha=0.9, edgecolor='#4682B4', boxstyle='round,pad=1'))

    # Final title
    fig.suptitle('CONTINUOUS DYNAMIC PERFORMANCE MAPPING\nGravitational Wave Pipeline Analysis',
                 fontsize=24, fontweight='bold', y=0.98)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# =============================================================================
# EXECUTE THE VISUALIZATION
# =============================================================================

# First run your comprehensive analysis to get 'results'
# Then call this function with the results object
if 'results' in globals() and results is not None:
    print("🎨 Creating continuous heatmaps from analysis results...")
    create_continuous_heatmaps(results)
    print("✅ Continuous visualization complete!")
else:
    print("❌ No results found. Please run your comprehensive analysis first!")
```

```python
# =============================================================================
# ENHANCED VISUAL HEATMAP - PATTERN FOCUSED (MODIFIED)
# =============================================================================

def create_enhanced_visual_heatmap(accuracy_matrix, error_matrix, success_matrix,
                                 snr_values, mass_values):
    """
    Create visually enhanced heatmap focused on pattern observation
    with improved visual hierarchy and pattern emphasis
    """
    # Set modern plotting style with enhanced readability
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.rcParams.update({
        'font.size': 11,
        'font.family': 'sans-serif',
        'axes.linewidth': 1.2,
        'grid.alpha': 0.25,
        'axes.edgecolor': '#2F4F4F',
        'axes.titleweight': 'bold'
    })

    # Create figure with better spacing and aspect ratio
    fig = plt.figure(figsize=(22, 16))
    gs = fig.add_gridspec(3, 3, hspace=0.4, wspace=0.3,
                         height_ratios=[1, 1, 0.7], width_ratios=[1, 1, 1])

    # Enhanced color palettes with better differentiation
    accuracy_cmap = 'RdYlGn'      # Red-Yellow-Green for accuracy
    error_cmap = 'magma_r'        # High-contrast reversed magma for errors
    success_cmap = 'Blues'        # Blues for success rate

    # Panel 1: Pipeline Accuracy Score (ENHANCED PATTERN VISIBILITY)
    ax1 = fig.add_subplot(gs[0, 0])
    sns.heatmap(accuracy_matrix,
                xticklabels=[f'{snr:.0f}' for snr in snr_values],
                yticklabels=[f'{mass:.0f}' for mass in mass_values],
                annot=False,
                cmap=accuracy_cmap,
                vmin=0, vmax=1,
                square=True,
                linewidths=0.8,
                linecolor='white',
                ax=ax1,
                cbar_kws={'label': 'Accuracy Score', 'shrink': 0.8})

    # Add contour lines for better pattern recognition
    contour = ax1.contour(accuracy_matrix, colors='black', linewidths=0.7, alpha=0.5)
    ax1.clabel(contour, inline=True, fontsize=9, fmt='%0.2f')

    ax1.set_title('Pipeline Accuracy Patterns\n(Higher = Better)', fontsize=13, pad=12)
    ax1.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax1.set_ylabel('Total Mass (M☉)', fontweight='bold')
    ax1.tick_params(axis='both', which='major', labelsize=10)

    # Panel 2: Relative Error (STRONGER CONTRAST)
    ax2 = fig.add_subplot(gs[0, 1])
    sns.heatmap(error_matrix,
                xticklabels=[f'{snr:.0f}' for snr in snr_values],
                yticklabels=[f'{mass:.0f}' for mass in mass_values],
                annot=False,
                cmap=error_cmap,
                vmin=0, vmax=60,
                square=True,
                linewidths=0.8,
                linecolor='white',
                ax=ax2,
                cbar_kws={'label': 'Relative Error (%)', 'shrink': 0.8})

    # Add threshold contour at 20% error level
    contour = ax2.contour(error_matrix, levels=[20], colors='white', linewidths=2)
    ax2.clabel(contour, inline=True, fontsize=10, fmt='%d%%')

    ax2.set_title('Estimation Error Patterns\n(Lower = Better)', fontsize=13, pad=12)
    ax2.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax2.set_ylabel('Total Mass (M☉)', fontweight='bold')
    ax2.tick_params(axis='both', which='major', labelsize=10)

    # Panel 3: Success Rate (OPTIMAL RANGE HIGHLIGHT)
    ax3 = fig.add_subplot(gs[0, 2])
    sns.heatmap(success_matrix,
                xticklabels=[f'{snr:.0f}' for snr in snr_values],
                yticklabels=[f'{mass:.0f}' for mass in mass_values],
                annot=False,
                cmap=success_cmap,
                vmin=0, vmax=1,
                square=True,
                linewidths=0.8,
                linecolor='white',
                ax=ax3,
                cbar_kws={'label': 'Success Rate', 'shrink': 0.8})

    # Highlight high-success region (>80%)
    success_mask = np.where(success_matrix > 0.8, 1, 0)
    ax3.contourf(success_mask, levels=[0.5, 1.5], colors='none',
                hatches=['', '////'], alpha=0)

    ax3.set_title('Analysis Success Patterns\n(Completion Rate)', fontsize=13, pad=12)
    ax3.set_xlabel('Signal-to-Noise Ratio', fontweight='bold')
    ax3.set_ylabel('Total Mass (M☉)', fontweight='bold')
    ax3.tick_params(axis='both', which='major', labelsize=10)

    # Panel 4: Enhanced Performance vs SNR (WITH CONFIDENCE BANDS)
    ax4 = fig.add_subplot(gs[1, 0])
    mean_accuracy_vs_snr = np.mean(accuracy_matrix, axis=0)
    std_accuracy_vs_snr = np.std(accuracy_matrix, axis=0)

    mean_success_vs_snr = np.mean(success_matrix, axis=0)
    std_success_vs_snr = np.std(success_matrix, axis=0)

    # Accuracy curve with confidence band
    ax4.plot(snr_values, mean_accuracy_vs_snr, 'o-', linewidth=3, markersize=8,
             color='#2E8B57', label='Accuracy', alpha=0.8)
    ax4.fill_between(snr_values,
                    mean_accuracy_vs_snr - std_accuracy_vs_snr,
                    mean_accuracy_vs_snr + std_accuracy_vs_snr,
                    alpha=0.2, color='#2E8B57')

    # Success rate curve with confidence band
    ax4.plot(snr_values, mean_success_vs_snr, 's-', linewidth=3, markersize=8,
             color='#DC143C', label='Success Rate', alpha=0.8)
    ax4.fill_between(snr_values,
                    mean_success_vs_snr - std_success_vs_snr,
                    mean_success_vs_snr + std_success_vs_snr,
                    alpha=0.2, color='#DC143C')

    # Add critical thresholds
    ax4.axvline(x=50, color='#FF8C00', linestyle='--', alpha=0.9, linewidth=2.5,
                label='Detection Threshold (SNR=50)')
    ax4.axvline(x=100, color='#32CD32', linestyle='--', alpha=0.9, linewidth=2.5,
                label='Optimal Threshold (SNR=100)')

    ax4.set_xlabel('Signal-to-Noise Ratio', fontweight='bold', fontsize=12)
    ax4.set_ylabel('Performance Score', fontweight='bold', fontsize=12)
    ax4.set_title('SNR Performance Scaling\n(Averaged over all masses)', fontsize=13)
    ax4.legend(loc='lower right', frameon=True, framealpha=0.9)
    ax4.grid(True, alpha=0.2, linestyle='-')
    ax4.set_xlim(snr_values[0]-5, snr_values[-1]+5)
    ax4.set_ylim(-0.05, 1.1)

    # Panel 5: Enhanced Performance vs Mass (WITH CONFIDENCE BANDS)
    ax5 = fig.add_subplot(gs[1, 1])
    mean_accuracy_vs_mass = np.mean(accuracy_matrix, axis=1)
    std_accuracy_vs_mass = np.std(accuracy_matrix, axis=1)

    mean_success_vs_mass = np.mean(success_matrix, axis=1)
    std_success_vs_mass = np.std(success_matrix, axis=1)

    # Accuracy curve with confidence band
    ax5.plot(mass_values, mean_accuracy_vs_mass, 'o-', linewidth=3, markersize=8,
             color='#4169E1', label='Accuracy', alpha=0.8)
    ax5.fill_between(mass_values,
                    mean_accuracy_vs_mass - std_accuracy_vs_mass,
                    mean_accuracy_vs_mass + std_accuracy_vs_mass,
                    alpha=0.2, color='#4169E1')

    # Success rate curve with confidence band
    ax5.plot(mass_values, mean_success_vs_mass, 's-', linewidth=3, markersize=8,
             color='#FF6347', label='Success Rate', alpha=0.8)
    ax5.fill_between(mass_values,
                    mean_success_vs_mass - std_success_vs_mass,
                    mean_success_vs_mass + std_success_vs_mass,
                    alpha=0.2, color='#FF6347')

    # Highlight optimal mass range with gradient
    ax5.axvspan(50, 70, alpha=0.15, color='gold', label='Optimal Mass Range')
    ax5.axvspan(70, 90, alpha=0.08, color='gold')

    ax5.set_xlabel('Total Mass (M☉)', fontweight='bold', fontsize=12)
    ax5.set_ylabel('Performance Score', fontweight='bold', fontsize=12)
    ax5.set_title('Mass Sensitivity Analysis\n(Averaged over all SNRs)', fontsize=13)
    ax5.legend(loc='lower left', frameon=True, framealpha=0.9)
    ax5.grid(True, alpha=0.2, linestyle='-')
    ax5.set_xlim(mass_values[0]-5, mass_values[-1]+5)
    ax5.set_ylim(-0.05, 1.1)

    # Panel 6: Enhanced Error Distribution (KDE + HISTOGRAM)
    ax6 = fig.add_subplot(gs[1, 2])
    successful_errors = error_matrix[success_matrix > 0]

    if len(successful_errors) > 0:
        # KDE plot for smooth distribution
        sns.kdeplot(successful_errors, ax=ax6, color='darkblue',
                   linewidth=2.5, alpha=0.7, label='Density')

        # Histogram with gradient coloring
        n, bins, patches = ax6.hist(successful_errors, bins=15, alpha=0.5,
                                   color='skyblue', edgecolor='navy',
                                   linewidth=1.2, density=True,
                                   label='Frequency')

        # Add statistics lines
        mean_err = np.mean(successful_errors)
        median_err = np.median(successful_errors)
        std_err = np.std(successful_errors)

        ax6.axvline(mean_err, color='red', linestyle='-', linewidth=2.5,
                   label=f'Mean: {mean_err:.1f}%')
        ax6.axvline(median_err, color='green', linestyle='-', linewidth=2.5,
                   label=f'Median: {median_err:.1f}%')
        ax6.axvline(mean_err - std_err, color='darkred', linestyle=':', linewidth=1.5)
        ax6.axvline(mean_err + std_err, color='darkred', linestyle=':', linewidth=1.5,
                   label=f'±1σ: {std_err:.1f}%')

        ax6.legend(frameon=True, framealpha=0.9)

    ax6.set_xlabel('Relative Error (%)', fontweight='bold', fontsize=12)
    ax6.set_ylabel('Density/Frequency', fontweight='bold', fontsize=12)
    ax6.set_title('Error Distribution\n(Successful Estimates Only)', fontsize=13)
    ax6.grid(True, alpha=0.2, linestyle='-')

    # Enhanced Summary Panel (DATA-DRIVEN INSIGHTS)
    ax7 = fig.add_subplot(gs[2, :])
    ax7.axis('off')

    # Calculate performance metrics
    overall_accuracy = np.nanmean(accuracy_matrix[success_matrix > 0])
    overall_error = np.nanmean(error_matrix[success_matrix > 0])
    overall_success = np.mean(success_matrix)

    # Identify high-performance regions
    high_perf_mask = (accuracy_matrix > 0.7) & (success_matrix > 0.8)
    optimal_snrs = []
    optimal_masses = []

    if np.any(high_perf_mask):
        high_perf_indices = np.where(high_perf_mask)
        for i, j in zip(*high_perf_indices):
            optimal_snrs.append(snr_values[j])
            optimal_masses.append(mass_values[i])

    # Calculate optimal parameter ranges
    opt_snr_range = (min(optimal_snrs), max(optimal_snrs)) if optimal_snrs else (0, 0)
    opt_mass_range = (min(optimal_masses), max(optimal_masses)) if optimal_masses else (0, 0)

    # Create data-driven summary
    summary_text = f"""
📊 PERFORMANCE SUMMARY: ZERO-CROSSING CHIRP MASS ESTIMATION

✅ Overall Success Rate: {overall_success:.1%} ({np.sum(success_matrix):.0f}/{success_matrix.size:.0f} tests)
🎯 Average Accuracy: {overall_accuracy:.3f}/1.000 (±{np.nanstd(accuracy_matrix[success_matrix>0]):.3f})
📏 Typical Error: {overall_error:.1f}% (±{np.nanstd(error_matrix[success_matrix>0]):.1f}%)

🔍 Key Performance Patterns:
   • SNR Threshold: Critical at SNR ≈ 50 (Success: {np.mean(success_matrix[:, snr_values <= 50]):.0%})
   • Optimal SNR: >100 (Accuracy: {np.nanmean(accuracy_matrix[:, snr_values > 100]):.1%})
   • Mass Range: 50-70 M☉ (Success: {np.nanmean(success_matrix[(mass_values >= 50) & (mass_values <= 70), :]):.0%})
   • High Mass Dropoff: >130 M☉ (Accuracy: {np.nanmean(accuracy_matrix[mass_values > 130, :]):.0%})

💡 Recommended Operating Parameters:
   • SNR Range: {opt_snr_range[0]:.0f}-{opt_snr_range[1]:.0f}
   • Mass Range: {opt_mass_range[0]:.0f}-{opt_mass_range[1]:.0f} M☉
   • Expected Accuracy: {np.nanmean(accuracy_matrix[high_perf_mask]):.1%} ± {np.nanstd(accuracy_matrix[high_perf_mask]):.1%}
"""

    # Create styled text box with improved layout
    ax7.text(0.02, 0.95, summary_text, transform=ax7.transAxes,
             fontsize=12, verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle='round,pad=0.8', facecolor='#F0F8FF',
                      edgecolor='#4682B4', alpha=0.95, linewidth=2))

    # Enhanced title
    fig.suptitle('GRAVITATIONAL WAVE PIPELINE PERFORMANCE: PATTERN ANALYSIS',
                 fontsize=20, fontweight='bold', y=0.99, color='#2F4F4F')

    # Professional footer
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M UTC")
    plt.figtext(0.99, 0.01, f"Analysis: {timestamp} | Zero-Crossing Method | Visual Pattern Focus",
                ha='right', fontsize=10, alpha=0.9, style='italic', color='#556B2F')

    plt.show()
    return fig

# =============================================================================
# CREATE PATTERN-FOCUSED VISUALIZATION (MODIFIED)
# =============================================================================

print("🎨 Creating enhanced pattern-focused visualization...")
print("Focus: Visual patterns over numerical precision")

enhanced_fig = create_enhanced_visual_heatmap(
    results['accuracy_matrix'],
    results['error_matrix'],
    results['success_matrix'],
    results['snr_values'],
    results['mass_values']
)

print("✨ Enhanced visualization complete!")
print("Key improvements:")
print("  • Added contour lines for pattern delineation")
print("  • Included confidence bands in trend plots")
print("  • Implemented gradient highlighting for optimal ranges")
print("  • Combined KDE and histogram in error distribution")
print("  • Data-driven summary with specific performance thresholds")
```
